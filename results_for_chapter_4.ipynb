## §4.3 -- COMO Function Figure (`fig:como_function`)

Plots $c_i = \max(0, \bar{s}_i - s_i)$ to illustrate the kink at $s_i = \bar{s}_i$.
The shaded region highlights where the individual uses social media less than
their peer average and thus experiences a cost of missing out.

```python
def plot_como_function(
    s_bar: float = 4.0,
    s_range: tuple[float, float] = (0, 8),
    figsize: tuple[float, float] = (8, 5),
    save_path: str | None = None,
) -> plt.Figure:
    """Plot the COMO function c = max(0, s̄ - s).

    Parameters
    ----------
    s_bar : float
        The peer average screen time (where the kink occurs)
    s_range : tuple
        (min, max) range for own screen time s
    figsize : tuple
        Figure size
    save_path : str, optional
        If provided, saves the figure to this path

    Returns
    -------
    fig : matplotlib.Figure
        The figure object
    """
    _set_thesis_style()

    s = np.linspace(s_range[0], s_range[1], 500)
    c = np.maximum(0, s_bar - s)

    fig, ax = plt.subplots(figsize=figsize)

    # Plot the COMO function
    ax.plot(
        s,
        c,
        color=THESIS_COLORS["steel_blue"],
        linewidth=2.5,
        label=r"$c_i = \max(0, \bar{s}_i - s_i)$",
    )

    # Add vertical line at s = s_bar
    ax.axvline(
        x=s_bar, color=THESIS_COLORS["gray"], linestyle="--", linewidth=1.5, alpha=0.7
    )

    # Add horizontal line at c = 0
    ax.axhline(
        y=0, color=THESIS_COLORS["light_gray"], linestyle="-", linewidth=0.8, alpha=0.5
    )

    # Shade the "missing out" region (where c > 0)
    s_missing = s[s < s_bar]
    c_missing = c[s < s_bar]
    ax.fill_between(
        s_missing,
        0,
        c_missing,
        alpha=0.15,
        color=THESIS_COLORS["muted_red"],
        label="Missing out region",
    )

    # Annotate the kink point
    ax.plot(s_bar, 0, "o", color=THESIS_COLORS["dark_gray"], markersize=8, zorder=5)
    ax.annotate(
        r"$s_i = \bar{s}_i$" + "\n(kink point)",
        xy=(s_bar, 0),
        xytext=(s_bar + 0.8, 1.2),
        fontsize=10,
        ha="left",
        arrowprops=dict(arrowstyle="->", color=THESIS_COLORS["dark_gray"], lw=1.2),
    )

    # Add region labels
    ax.text(
        s_bar / 2,
        s_bar / 4 + 0.3,
        r"$c_i > 0$" + "\n(missing out)",
        fontsize=10,
        ha="center",
        va="center",
        color=THESIS_COLORS["muted_red"],
        fontweight="medium",
    )
    ax.text(
        (s_bar + s_range[1]) / 2,
        0.3,
        r"$c_i = 0$" + "\n(no cost)",
        fontsize=10,
        ha="center",
        va="bottom",
        color=THESIS_COLORS["sea_green"],
        fontweight="medium",
    )

    # Labels
    ax.set_xlabel(r"Own screen time $s_i$ (hours/day)")
    ax.set_ylabel(r"COMO term $c_i$")
    ax.set_title("The Cost-of-Missing-Out (COMO) Function", fontweight="medium")

    # Set axis limits with padding
    ax.set_xlim(s_range[0] - 0.2, s_range[1] + 0.2)
    ax.set_ylim(-0.3, s_bar + 0.5)

    # Add subtle grid
    ax.yaxis.grid(True, linestyle="-", alpha=0.3, color="gray")
    ax.set_axisbelow(True)

    ax.legend(loc="upper right", framealpha=0.95, edgecolor="none")

    fig.tight_layout()

    if save_path:
        fig.savefig(save_path, dpi=300, bbox_inches="tight")
        print(f"Saved figure to {save_path}")

    return fig


fig1 = plot_como_function(
    s_bar=4.0,
    save_path=str(FIGURES_DIR / "como_function.pdf"),
)
plt.show()
```

## §4.7 -- Illustrative Example: Computed Values

Compute all numerical values for Section 4.7 of Chapter 4. This cell is the
single source of truth for the numbers quoted in the thesis text and used in
the network diagram below.

**Parameters** (matching §4.7 text):
- SMU: $\beta_0 = 2.5$, $\beta_1 = 0.2$, $\sigma_u = 0.5$
- WB: $\gamma_0 = 7$, $\gamma_1 = -0.3$, $\gamma_2 = -0.5$, $\gamma_3 = 0.2$, $\sigma_\varepsilon = 0.4$

**Computation steps:**
1. Solve the SMU reduced form $\mathbf{s} = (I - \beta_1 G)^{-1}(\beta_0 \mathbf{1} + \mathbf{u})$
2. Peer averages $G\mathbf{s}$
3. COMO: $c_i = \max(0, \bar{s}_i - s_i)$
4. Solve the WB reduced form $\mathbf{y} = (I - \gamma_3 G)^{-1}(\gamma_0 \mathbf{1} + \gamma_1 \mathbf{s} + \gamma_2 \mathbf{c} + \boldsymbol{\varepsilon})$
5. Friends-of-friends matrix $G^2$ for the identification illustration

```python
# --- Parameters (matching Section 4.7 text) ---
beta0, beta1 = 2.5, 0.2
# sigma_u = 0.5 (not used; error terms are fixed for the illustration)
u = np.array([0.2, -0.3, 0.1, 0.4])

gamma0, gamma1, gamma2, gamma3 = 7.0, -0.3, -0.5, 0.2
# sigma_eps = 0.4 (not used; error terms are fixed for the illustration)
eps = np.array([0.1, -0.2, 0.3, -0.1])

# Row-normalized adjacency matrix
G = np.array(
    [
        [0, 0.5, 0.5, 0],
        [0.5, 0, 0, 0.5],
        [1, 0, 0, 0],
        [0, 1, 0, 0],
    ]
)

I = np.eye(4)

# --- Step 1: SMU via reduced form ---
A_s = I - beta1 * G
s = np.linalg.solve(A_s, beta0 * np.ones(4) + u)

# --- Step 2: Peer averages ---
Gs = G @ s

# --- Step 3: COMO ---
c = np.maximum(0, Gs - s)

# --- Step 4: Well-being via reduced form ---
X_gamma = gamma0 * np.ones(4) + gamma1 * s + gamma2 * c
A_y = I - gamma3 * G
y = np.linalg.solve(A_y, X_gamma + eps)

# --- Step 5: G^2 (friends-of-friends) ---
G2 = G @ G

# --- Print results ---
print("=" * 60)
print("Section 4.7 - Computed Values")
print("=" * 60)

print(f"\ns  = ({', '.join(f'{v:.2f}' for v in s)})")
print(f"Gs = ({', '.join(f'{v:.2f}' for v in Gs)})")

print("\nCOMO:")
for i in range(4):
    print(
        f"  c_{i+1} = max(0, {Gs[i]:.2f} - {s[i]:.2f}) = max(0, {Gs[i]-s[i]:.2f}) = {c[i]:.2f}"
    )

print(f"\ny  = ({', '.join(f'{v:.2f}' for v in y)})")

# COMO contribution for individual with largest COMO
idx = np.argmax(c)
contrib = gamma2 * c[idx]
print(
    f"\nCOMO contribution for individual {idx+1}: "
    f"gamma2 * c_{idx+1} = {gamma2} * {c[idx]:.2f} = {contrib:.2f}"
)

print(f"\nLowest WB: individual {np.argmin(y)+1} (y = {y[np.argmin(y)]:.2f})")

# Verify self-consistency
print("\n--- Self-consistency check ---")
s_check = beta0 + beta1 * (G @ s) + u
y_check = gamma0 + gamma1 * s + gamma2 * c + gamma3 * (G @ y) + eps
print(f"max |s - s_check| = {np.max(np.abs(s - s_check)):.2e}")
print(f"max |y - y_check| = {np.max(np.abs(y - y_check)):.2e}")

print("\nG^2 =")
print(np.array2string(G2, formatter={"float_kind": lambda x: f"{x:.2f}"}))

# Store computed values as dicts for the network diagram
screen_times = {i+1: s[i] for i in range(4)}
peer_averages = {i+1: Gs[i] for i in range(4)}
como_values = {i+1: c[i] for i in range(4)}
wellbeing = {i+1: y[i] for i in range(4)}
```

## §4.7 -- Network Diagram (`fig:network_example`)

Visualize the four-node network from the illustrative example. Node colors
indicate whether the individual experiences COMO (red) or not (blue).
Annotation boxes show $s_i$, $\bar{s}_i$, $c_i$, and $y_i$ for each node,
using the values computed in the cell above.

```python
def plot_network_example(
    screen_times: dict,
    peer_averages: dict,
    como_values: dict,
    wellbeing: dict,
    figsize: tuple[float, float] = (10, 6),
    save_path: str | None = None,
) -> plt.Figure:
    """Plot the 4-node network from the illustrative example (Section 4.7).

    Network structure:
    - Node 1 connects to nodes 2 and 3
    - Node 2 connects to nodes 1 and 4
    - Node 3 connects to node 1 only
    - Node 4 connects to node 2 only

    Parameters
    ----------
    screen_times : dict
        {node: s_i} screen time values
    peer_averages : dict
        {node: s_bar_i} peer average values
    como_values : dict
        {node: c_i} COMO values
    wellbeing : dict
        {node: y_i} well-being values
    figsize : tuple
        Figure size
    save_path : str, optional
        If provided, saves the figure to this path

    Returns
    -------
    fig : matplotlib.Figure
        The figure object
    """
    _set_thesis_style()

    fig, ax = plt.subplots(figsize=figsize)

    # Node positions (arranged to show structure clearly)
    positions = {
        1: (0.3, 0.5),
        2: (0.7, 0.5),
        3: (0.1, 0.2),
        4: (0.9, 0.2),
    }

    # Edges (undirected)
    edges = [(1, 2), (1, 3), (2, 4)]

    # Draw edges first (so nodes appear on top)
    for i, j in edges:
        x1, y1 = positions[i]
        x2, y2 = positions[j]
        ax.plot(
            [x1, x2],
            [y1, y2],
            color=THESIS_COLORS["gray"],
            linewidth=2.5,
            zorder=1,
            solid_capstyle="round",
        )

    # Draw nodes
    node_radius = 0.065
    for node, (x, y_pos) in positions.items():
        # Color based on COMO (red if experiencing COMO, blue otherwise)
        if como_values[node] > 0:
            node_color = THESIS_COLORS["muted_red"]
            edge_color = "#8B2323"  # Darker red
        else:
            node_color = THESIS_COLORS["steel_blue"]
            edge_color = "#2F5496"  # Darker blue

        circle = plt.Circle(
            (x, y_pos),
            node_radius,
            color=node_color,
            ec=edge_color,
            linewidth=2.5,
            zorder=3,
        )
        ax.add_patch(circle)

        # Node label (number)
        ax.text(
            x,
            y_pos,
            str(node),
            fontsize=16,
            fontweight="bold",
            ha="center",
            va="center",
            color="white",
            zorder=4,
        )

    # Add data annotations for each node
    annotation_offsets = {
        1: (-0.12, 0.12),
        2: (0.12, 0.12),
        3: (-0.12, -0.08),
        4: (0.12, -0.08),
    }

    for node, (x, y_pos) in positions.items():
        ox, oy = annotation_offsets[node]

        # Build annotation text
        text_lines = [
            f"$s_{node} = {screen_times[node]:.2f}$",
            f"$\\bar{{s}}_{node} = {peer_averages[node]:.2f}$",
        ]
        if como_values[node] > 0:
            text_lines.append(f"$c_{node} = {como_values[node]:.2f}$")
        else:
            text_lines.append(f"$c_{node} = 0$")
        text_lines.append(f"$y_{node} = {wellbeing[node]:.2f}$")

        annotation_text = "\n".join(text_lines)

        # Highlight box color
        if como_values[node] > 0:
            box_color = "#FFE4E1"  # Light red
        else:
            box_color = "#E8F4FD"  # Light blue

        ax.annotate(
            annotation_text,
            xy=(x, y_pos),
            xytext=(x + ox, y_pos + oy),
            fontsize=9,
            ha="center" if ox == 0 else ("right" if ox < 0 else "left"),
            va="center",
            bbox=dict(
                boxstyle="round,pad=0.4",
                facecolor=box_color,
                edgecolor=THESIS_COLORS["light_gray"],
                linewidth=1,
            ),
            zorder=5,
        )

    # Add legend explaining colors
    legend_elements = [
        mpatches.Patch(
            facecolor=THESIS_COLORS["steel_blue"],
            edgecolor="#2F5496",
            linewidth=1.5,
            label=r"$c_i = 0$ (no missing out)",
        ),
        mpatches.Patch(
            facecolor=THESIS_COLORS["muted_red"],
            edgecolor="#8B2323",
            linewidth=1.5,
            label=r"$c_i > 0$ (experiencing COMO)",
        ),
    ]
    ax.legend(
        handles=legend_elements,
        loc="lower center",
        ncol=2,
        framealpha=0.95,
        edgecolor="none",
        fontsize=10,
    )

    # Add title
    ax.set_title("Four-Node Network Example", fontweight="medium", fontsize=14)

    # Set axis properties
    ax.set_xlim(-0.05, 1.05)
    ax.set_ylim(-0.05, 0.85)
    ax.set_aspect("equal")
    ax.axis("off")

    fig.tight_layout()

    if save_path:
        fig.savefig(save_path, dpi=300, bbox_inches="tight")
        print(f"Saved figure to {save_path}")

    return fig


fig2 = plot_network_example(
    screen_times=screen_times,
    peer_averages=peer_averages,
    como_values=como_values,
    wellbeing=wellbeing,
    save_path=str(FIGURES_DIR / "network_example_chapter_4.pdf"),
)
plt.show()
```
