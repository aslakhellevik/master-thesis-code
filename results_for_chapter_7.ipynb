## Key implementation details

- Concentrated FIML in 5 parameters $(\beta_1, \beta_2, \gamma_1, \gamma_2, \gamma_3)$; $\beta_0, \gamma_0, \sigma_u^2, \sigma_\varepsilon^2$ are profiled out analytically.
- Block Jacobian determinant identity $\det(J) = \det(A_y A_s - \beta_2 \Gamma)$, avoiding the dense $A_s^{-1}$ -- see the markdown cell preceding `fiml_neg_loglik_full` for the derivation.
- Sparse LU (SuperLU) for log-determinants and equilibrium solves.
- Warm-start from recursive ML estimates for robust initialization of the 5D optimizer.
- Numerical Hessian for SEs (four-point central difference).

## Core Functions: Sparse Utilities

```python
def generate_network_sparse(n, p, rng=None):
    """Generate sparse Erdos-Renyi network with row-normalized adjacency matrix.

    Redraws until the network contains no isolated nodes, so that n is fixed
    and G is strictly row-stochastic. Matches the approach in Chapter 6.
    """
    if rng is None:
        rng = np.random.default_rng()
    while True:
        upper = rng.random((n, n)) < p
        upper = np.triu(upper, k=1)
        adj = upper | upper.T
        degrees = adj.sum(axis=1)
        if np.all(degrees > 0):
            break

    n = adj.shape[0]
    rows, cols, vals = [], [], []
    for i in range(n):
        neighbors = np.where(adj[i])[0]
        w = 1.0 / len(neighbors)
        for j in neighbors:
            rows.append(i)
            cols.append(j)
            vals.append(w)
    return sp.csr_matrix((vals, (rows, cols)), shape=(n, n))


def _perm_sign(perm):
    """Compute sign of a permutation (+1 or -1) by counting transpositions."""
    n = len(perm)
    visited = np.zeros(n, dtype=bool)
    n_swaps = 0
    for i in range(n):
        if not visited[i]:
            j = i
            cycle_len = 0
            while not visited[j]:
                visited[j] = True
                j = perm[j]
                cycle_len += 1
            n_swaps += cycle_len - 1
    return 1 if n_swaps % 2 == 0 else -1


def sparse_slogdet(M_sparse):
    """
    Compute (sign, log|det|) for a sparse matrix via SuperLU factorization.

    Equivalent to numpy.linalg.slogdet but exploits sparsity structure.
    Uses Pr @ M @ Pc = L @ U, so det(M) = det(Pr)^{-1} det(L) det(U) det(Pc)^{-1}.
    Since L has unit diagonal, det(L)=1, and det(M) = sign_perms * prod(diag(U)).

    Returns (0, -inf) for singular matrices (splu raises RuntimeError).
    """
    try:
        lu = splu(sp.csc_matrix(M_sparse))
    except RuntimeError:
        return 0, -np.inf
    diag_U = lu.U.diagonal().copy()
    abs_diag = np.abs(diag_U)
    if np.any(abs_diag == 0):
        return 0, -np.inf
    logdet = np.sum(np.log(abs_diag))
    n_neg = np.sum(diag_U < 0)
    sign_U = 1 if n_neg % 2 == 0 else -1
    sign = _perm_sign(lu.perm_r) * sign_U * _perm_sign(lu.perm_c)
    return sign, logdet
```

## Simultaneous Model: Fixed-Point Data Generation

```python
def simulate_simultaneous_data(G_sp, beta0, beta1, beta2, sigma_u_sq,
                                gamma0, gamma1, gamma2, gamma3, sigma_eps_sq,
                                rng=None, max_iter=200, tol=1e-6):
    """
    Simulate data from the simultaneous COMO model via fixed-point iteration.

    Uses pre-computed sparse LU factorizations of A_s and A_y for fast solves.

    SMU:  s_i = beta0 + beta1 * s_bar_i + beta2 * y_i + u_i
    WB:   y_i = gamma0 + gamma1 * s_i + gamma2 * c_i + gamma3 * y_bar_i + eps_i

    Returns (s, y, converged, n_iter)
    """
    if rng is None:
        rng = np.random.default_rng()
    n = G_sp.shape[0]
    I_sp = sp.eye(n, format='csc')
    ones = np.ones(n)

    A_s_csc = (I_sp - beta1 * G_sp).tocsc()
    A_y_csc = (I_sp - gamma3 * G_sp).tocsc()
    lu_As = splu(A_s_csc)
    lu_Ay = splu(A_y_csc)

    u = rng.normal(0, np.sqrt(sigma_u_sq), n)
    eps = rng.normal(0, np.sqrt(sigma_eps_sq), n)
    # Initial guess for fixed-point iteration (overwritten upon convergence)
    y = gamma0 * ones + rng.normal(0, np.sqrt(sigma_eps_sq), n)

    converged = False
    for it in range(max_iter):
        s_new = lu_As.solve(beta0 * ones + beta2 * y + u)
        s_bar = G_sp @ s_new
        c = np.maximum(0, s_bar - s_new)
        y_new = lu_Ay.solve(gamma0 * ones + gamma1 * s_new + gamma2 * c + eps)

        if np.max(np.abs(y_new - y)) < tol:
            converged = True
            y = y_new
            s = s_new
            break
        y = y_new
    else:
        s = s_new

    return s, y, converged, it + 1
```

## FIML Likelihood Functions (Sparse)

**Key optimization:** The Jacobian of the simultaneous system has block structure
$$
J = \begin{pmatrix} A_s & -\beta_2 I \\ -\Gamma & A_y \end{pmatrix}
$$
The standard approach computes $\det(J) = \det(A_s) \det(A_y - \beta_2 \Gamma A_s^{-1})$,
requiring the dense inverse $A_s^{-1}$.

Instead, we use the identity
$$
\det(J) = \det(A_y A_s - \beta_2 \Gamma)
$$
which follows from $A_y - \beta_2 \Gamma A_s^{-1} = (A_y A_s - \beta_2 \Gamma) A_s^{-1}$.

Since $A_y$, $A_s$, and $\Gamma$ are all sparse, $M = A_y A_s - \beta_2 \Gamma$ is sparse,
and its log-determinant can be computed efficiently via sparse LU (SuperLU).

```python
def fiml_neg_loglik_full(params, s, y, G_sp, I_sp):
    """
    Full FIML negative log-likelihood for the simultaneous model (sparse).

    params = [beta0, beta1, beta2, gamma0, gamma1, gamma2, gamma3, sigma_u_sq, sigma_eps_sq]

    Uses det(J) = det(A_y @ A_s - beta2 * Gamma) to avoid dense inv(A_s).
    """
    beta0, beta1, beta2, gamma0, gamma1, gamma2, gamma3, sigma_u_sq, sigma_eps_sq = params

    if abs(beta1) >= 0.99 or abs(gamma3) >= 0.99 or sigma_u_sq <= 0 or sigma_eps_sq <= 0:
        return np.inf

    n = len(s)
    ones = np.ones(n)

    A_s = I_sp - beta1 * G_sp
    A_y = I_sp - gamma3 * G_sp

    # Structural residuals (sparse matvec -> dense vectors)
    u = A_s @ s - beta0 * ones - beta2 * y
    s_bar = G_sp @ s
    c = np.maximum(0, s_bar - s)
    eps_vec = A_y @ y - gamma0 * ones - gamma1 * s - gamma2 * c

    # Sparse Gamma matrix
    d = (s_bar > s).astype(float)
    D_sp = sp.diags(d)
    Gamma_sp = gamma1 * I_sp + gamma2 * D_sp @ (G_sp - I_sp)

    # det(J) = det(A_y @ A_s - beta2 * Gamma)  [all sparse]
    M = A_y @ A_s - beta2 * Gamma_sp

    sign, logdet = sparse_slogdet(M)
    if sign == 0:
        return np.inf

    nll = (-logdet
           + (n / 2) * np.log(sigma_u_sq) + (n / 2) * np.log(sigma_eps_sq)
           + np.dot(u, u) / (2 * sigma_u_sq) + np.dot(eps_vec, eps_vec) / (2 * sigma_eps_sq))

    return nll


def fiml_neg_loglik_concentrated(theta5, s, y, G_sp, I_sp):
    """
    Concentrated FIML negative log-likelihood (sparse).

    theta5 = [beta1, beta2, gamma1, gamma2, gamma3]
    Profiles out: beta0, gamma0, sigma_u_sq, sigma_eps_sq
    """
    beta1, beta2, gamma1, gamma2, gamma3 = theta5

    if abs(beta1) >= 0.99 or abs(gamma3) >= 0.99:
        return np.inf

    n = len(s)
    ones = np.ones(n)

    A_s = I_sp - beta1 * G_sp
    A_y = I_sp - gamma3 * G_sp

    # Profile beta0
    As_s = A_s @ s
    beta0_hat = float(np.mean(As_s - beta2 * y))
    u = As_s - beta0_hat * ones - beta2 * y

    # Profile gamma0
    s_bar = G_sp @ s
    c = np.maximum(0, s_bar - s)
    Ay_y = A_y @ y
    gamma0_hat = float(np.mean(Ay_y - gamma1 * s - gamma2 * c))
    eps_vec = Ay_y - gamma0_hat * ones - gamma1 * s - gamma2 * c

    # Profile variances
    sigma_u_sq_hat = float(np.dot(u, u) / n)
    sigma_eps_sq_hat = float(np.dot(eps_vec, eps_vec) / n)

    if sigma_u_sq_hat <= 0 or sigma_eps_sq_hat <= 0:
        return np.inf

    # Sparse Gamma and Jacobian determinant
    d = (s_bar > s).astype(float)
    D_sp = sp.diags(d)
    Gamma_sp = gamma1 * I_sp + gamma2 * D_sp @ (G_sp - I_sp)
    M = A_y @ A_s - beta2 * Gamma_sp

    sign, logdet = sparse_slogdet(M)
    if sign == 0:
        return np.inf

    nll = -logdet + (n / 2) * np.log(sigma_u_sq_hat) + (n / 2) * np.log(sigma_eps_sq_hat) + n
    return nll
```

## FIML Estimation (Warm-Started, Sparse)

```python
def estimate_fiml(s, y, G_sp, I_sp, warm_start=None, compute_se=True):
    """
    Estimate the simultaneous model by FIML.

    Uses concentrated likelihood (5D) then recovers all 9 parameters.
    Warm-started from recursive ML estimates for robust initialization.
    Optionally computes numerical Hessian-based SEs (set compute_se=False to skip).

    Parameters
    ----------
    warm_start : array-like of length 5, optional
        Starting point [beta1, beta2, gamma1, gamma2, gamma3] for the optimizer,
        e.g. from recursive ML estimates. Falls back to zeros if not provided.
    """
    n = len(s)
    ones = np.ones(n)

    bounds = [(-0.98, 0.98), (-2.0, 2.0), (-5.0, 5.0), (-5.0, 5.0), (-0.98, 0.98)]

    # Starting point: warm start from recursive ML, or zeros as fallback
    if warm_start is not None:
        x0 = np.asarray(warm_start, dtype=float).copy()
    else:
        x0 = np.zeros(5)

    lower = np.array([b[0] for b in bounds], dtype=float)
    upper = np.array([b[1] for b in bounds], dtype=float)
    x0 = np.clip(x0, lower + 1e-4, upper - 1e-4)

    # Optimize concentrated likelihood
    res = minimize(
        fiml_neg_loglik_concentrated, x0,
        args=(s, y, G_sp, I_sp),
        method='L-BFGS-B', bounds=bounds,
        options={'maxiter': 250, 'ftol': 1e-8}
    )
    if (not res.success) or (not np.isfinite(res.fun)):
        # Fallback to Nelder-Mead
        res2 = minimize(
            fiml_neg_loglik_concentrated, x0,
            args=(s, y, G_sp, I_sp),
            method='Nelder-Mead',
            options={'maxiter': 2000, 'xatol': 1e-8, 'fatol': 1e-8}
        )
        if np.isfinite(res2.fun) and (not np.isfinite(res.fun) or res2.fun < res.fun):
            res = res2

    param_names = ['beta0', 'beta1', 'beta2', 'gamma0', 'gamma1', 'gamma2', 'gamma3',
                   'sigma_u_sq', 'sigma_eps_sq']

    if not np.isfinite(res.fun):
        result = {k: np.nan for k in param_names}
        for k in param_names:
            result[f'se_{k}'] = np.nan
        result['success'] = False
        result['nll'] = np.nan
        return result

    beta1_hat, beta2_hat, gamma1_hat, gamma2_hat, gamma3_hat = res.x

    # Recover profiled parameters
    A_s = I_sp - beta1_hat * G_sp
    A_y = I_sp - gamma3_hat * G_sp
    As_s = A_s @ s
    beta0_hat = float(np.mean(As_s - beta2_hat * y))
    u = As_s - beta0_hat * ones - beta2_hat * y

    s_bar = G_sp @ s
    c = np.maximum(0, s_bar - s)
    Ay_y = A_y @ y
    gamma0_hat = float(np.mean(Ay_y - gamma1_hat * s - gamma2_hat * c))
    eps_vec = Ay_y - gamma0_hat * ones - gamma1_hat * s - gamma2_hat * c

    sigma_u_sq_hat = float(np.dot(u, u) / n)
    sigma_eps_sq_hat = float(np.dot(eps_vec, eps_vec) / n)

    theta_full = np.array([beta0_hat, beta1_hat, beta2_hat,
                           gamma0_hat, gamma1_hat, gamma2_hat, gamma3_hat,
                           sigma_u_sq_hat, sigma_eps_sq_hat])

    # Standard errors via numerical Hessian (optional)
    se = np.full(9, np.nan)
    if compute_se:
        try:
            H = numerical_hessian_full(
                lambda p: fiml_neg_loglik_full(p, s, y, G_sp, I_sp),
                theta_full, eps_h=1e-5
            )
            cov = dense_inv(H)
            diag_cov = np.diag(cov)
            se = np.sqrt(np.maximum(diag_cov, 0))
            se[diag_cov < 0] = np.nan
        except Exception:
            pass

    result = {name: val for name, val in zip(param_names, theta_full)}
    for i, name in enumerate(param_names):
        result[f'se_{name}'] = se[i]
    result['success'] = bool(res.success and np.isfinite(res.fun))
    result['nll'] = float(res.fun)

    return result


def numerical_hessian_full(func, params, eps_h=1e-5):
    """Four-point central difference Hessian."""
    p = len(params)
    H = np.zeros((p, p))
    for i in range(p):
        for j in range(i, p):
            pp = params.copy(); pp[i] += eps_h; pp[j] += eps_h
            pm = params.copy(); pm[i] += eps_h; pm[j] -= eps_h
            mp = params.copy(); mp[i] -= eps_h; mp[j] += eps_h
            mm = params.copy(); mm[i] -= eps_h; mm[j] -= eps_h
            H[i, j] = (func(pp) - func(pm) - func(mp) + func(mm)) / (4 * eps_h**2)
            H[j, i] = H[i, j]
    return H
```

```python
# Recursive ML estimators (used to warm-start FIML, and reused in the bias comparison cells below)

def estimate_screen_ml_sparse(s, G_sp):
    """ML estimation of the SMU equation (recursive model, sparse)."""
    n = len(s)
    I_sp_loc = sp.eye(n, format='csr')

    def neg_conc_loglik(beta1):
        if abs(beta1) >= 0.99:
            return np.inf
        A = I_sp_loc - beta1 * G_sp
        As = A @ s
        beta0_hat = As.mean()
        resid = As - beta0_hat
        sse = float(np.dot(resid, resid))
        if sse <= 0:
            return np.inf
        sign, logdet = sparse_slogdet(A)
        if sign <= 0:
            return np.inf
        return -logdet + (n / 2) * np.log(sse / n)

    res = minimize_scalar(neg_conc_loglik, bounds=(-0.98, 0.98), method='bounded')
    beta1_hat = res.x
    A = I_sp_loc - beta1_hat * G_sp
    As = A @ s
    beta0_hat = float(As.mean())
    resid = As - beta0_hat
    sigma_u2_hat = float(np.dot(resid, resid) / n)
    return {'beta0': beta0_hat, 'beta1': beta1_hat, 'sigma_u_sq': sigma_u2_hat}


def estimate_wb_ml_sparse(y, s, G_sp):
    """ML estimation of the WB equation (recursive model, sparse)."""
    n = len(y)
    I_sp_loc = sp.eye(n, format='csr')
    s_bar = G_sp @ s
    c = np.maximum(0, s_bar - s)
    X = np.column_stack([np.ones(n), s, c])

    def neg_conc_loglik(gamma3):
        if abs(gamma3) >= 0.99:
            return np.inf
        A = I_sp_loc - gamma3 * G_sp
        Ay = A @ y
        coef, _, _, _ = np.linalg.lstsq(X, Ay, rcond=None)
        resid = Ay - X @ coef
        sse = float(np.dot(resid, resid))
        if sse <= 0:
            return np.inf
        sign, logdet = sparse_slogdet(A)
        if sign <= 0:
            return np.inf
        return -logdet + (n / 2) * np.log(sse / n)

    res = minimize_scalar(neg_conc_loglik, bounds=(-0.98, 0.98), method='bounded')
    gamma3_hat = res.x
    A = I_sp_loc - gamma3_hat * G_sp
    Ay = A @ y
    coef, _, _, _ = np.linalg.lstsq(X, Ay, rcond=None)
    resid = Ay - X @ coef
    sigma_eps2_hat = float(np.dot(resid, resid) / n)
    gamma0_hat, gamma1_hat, gamma2_hat = coef
    return {'gamma0': float(gamma0_hat), 'gamma1': float(gamma1_hat),
            'gamma2': float(gamma2_hat), 'gamma3': gamma3_hat,
            'sigma_eps_sq': sigma_eps2_hat}


def build_warm_start(s, y, G_sp):
    """Build a warm start for FIML from recursive ML estimates.
    
    Returns [beta1, beta2, gamma1, gamma2, gamma3] with beta2=0 (recursive assumption).
    """
    rec_s = estimate_screen_ml_sparse(s, G_sp)
    rec_w = estimate_wb_ml_sparse(y, s, G_sp)
    return np.array([rec_s['beta1'], 0.0, rec_w['gamma1'], rec_w['gamma2'], rec_w['gamma3']])


print("Recursive ML estimators and warm-start builder defined.")
print("These are used to initialize FIML optimization in the MC study.")
```

## Monte Carlo Study (`tab:mc_simul_results`, `tab:mc_simul_rejection`)

```python
def run_one_replication(n, p, true, seed, compute_se=True):
    """
    Run one MC replication for the simultaneous model.

    Generates a new network each replication (matching Ch.6 design).
    Uses recursive ML estimates as a warm start for FIML.
    Returns a dict with per-replication estimates and diagnostics.
    """
    param_names = ['beta0', 'beta1', 'beta2', 'gamma0', 'gamma1', 'gamma2', 'gamma3',
                   'sigma_u_sq', 'sigma_eps_sq']

    out = {
        'estimates': {k: np.nan for k in param_names},
        'std_errors': {k: np.nan for k in param_names},
        'fp_converged': False,
        'fp_iters': np.nan,
        'optim_success': False,
    }

    try:
        rng = np.random.default_rng(seed)

        # Generate new network for this replication
        G_sp = generate_network_sparse(n, p, rng)
        n_actual = G_sp.shape[0]
        I_sp = sp.eye(n_actual, format='csr')

        s, y, conv, n_iter = simulate_simultaneous_data(G_sp, **true, rng=rng)
        out['fp_converged'] = bool(conv)
        out['fp_iters'] = float(n_iter)

        if not conv:
            return out

        # Warm start from recursive ML estimates
        ws = build_warm_start(s, y, G_sp)

        res = estimate_fiml(s, y, G_sp, I_sp, warm_start=ws,
                           compute_se=compute_se)
        for k in param_names:
            out['estimates'][k] = res[k]
            out['std_errors'][k] = res[f'se_{k}']
        out['optim_success'] = bool(res['success'])

    except Exception:
        pass

    return out


def _aggregate_replication_results(rep_results, param_names):
    estimates = {k: np.array([rr['estimates'][k] for rr in rep_results], dtype=float) for k in param_names}
    std_errors = {k: np.array([rr['std_errors'][k] for rr in rep_results], dtype=float) for k in param_names}
    fp_converged = np.array([rr['fp_converged'] for rr in rep_results], dtype=bool)
    fp_iters = np.array([rr['fp_iters'] for rr in rep_results], dtype=float)
    optim_success = np.array([rr['optim_success'] for rr in rep_results], dtype=bool)

    return {
        'estimates': estimates,
        'std_errors': std_errors,
        'fp_converged': fp_converged,
        'fp_iters': fp_iters,
        'optim_success': optim_success,
    }


def run_simultaneous_mc_parallel(n, p, true, R, seed=42, n_jobs=None,
                                  compute_se=True):
    """
    Run MC for simultaneous model with new network per replication in parallel.
    """
    param_names = ['beta0', 'beta1', 'beta2', 'gamma0', 'gamma1', 'gamma2', 'gamma3',
                   'sigma_u_sq', 'sigma_eps_sq']

    seed_seq = np.random.SeedSequence(seed)
    child_seqs = seed_seq.spawn(R)
    rep_seeds = [int(cs.generate_state(1, dtype=np.uint32)[0]) for cs in child_seqs]

    if n_jobs is None:
        n_jobs = -1

    with parallel_config(backend='loky', inner_max_num_threads=1):
        rep_results = Parallel(n_jobs=n_jobs)(
            delayed(run_one_replication)(n, p, true, rep_seed,
                                         compute_se=compute_se)
            for rep_seed in rep_seeds
        )

    return _aggregate_replication_results(rep_results, param_names)
```

```python
print(f"Running simultaneous MC: n={N_SIMUL}, p={P_SIMUL}, R={R_SIMUL}")
print("Using warm-started FIML (single start from recursive ML estimates).")
print("New network generated each replication (matching Ch.6 design).")
print("This will take a while...")
t_start = time.time()
mc_simul = run_simultaneous_mc_parallel(
    N_SIMUL, P_SIMUL, TRUE_SIMUL, R_SIMUL, seed=RNG_SEED, n_jobs=-1,
    compute_se=True
)
t_mc = time.time() - t_start

print(f"\nDone in {t_mc:.0f}s ({t_mc/60:.1f} min)")
print(f"  Per-replication average: {t_mc/R_SIMUL:.2f}s (wall-clock / R, parallelized)")
print(f"\nFixed-point convergence rate: {mc_simul['fp_converged'].mean()*100:.1f}%")
valid_iters = mc_simul['fp_iters'][np.isfinite(mc_simul['fp_iters'])]
print(f"Mean FP iterations: {valid_iters.mean():.1f}")
print(f"Optimization success rate: {mc_simul['optim_success'].mean()*100:.1f}%")
```

## Wald Test for Feedback: Power Analysis

Test $H_0: \beta_2 = 0$ (no feedback) using the FIML estimates.
Since the true $\beta_2 = -0.15$, this gives the empirical power of the test.
The test statistic is $z = \hat{\beta}_2 / \hat{\sigma}_{\beta_2}$,
which is asymptotically $N(0,1)$ under $H_0$.

```python
# Wald test for H0: beta2 = 0 (no feedback)
# Since true beta2 = -0.15, this is a power analysis
est_b2 = mc_simul['estimates']['beta2']
se_b2 = mc_simul['std_errors']['beta2']
mask = np.isfinite(est_b2) & np.isfinite(se_b2) & (se_b2 > 0)
z_feedback = est_b2[mask] / se_b2[mask]  # test statistic under H0: beta2 = 0
p_feedback = 2 * (1 - stats.norm.cdf(np.abs(z_feedback)))

power_05 = np.mean(p_feedback < 0.05)
power_01 = np.mean(p_feedback < 0.01)
power_10 = np.mean(p_feedback < 0.10)

true_b2 = TRUE_SIMUL['beta2']
print('Wald Test for H0: beta2 = 0 (no feedback from WB to SMU)')
print('=' * 60)
print(f'True beta2 = {true_b2}')
print(f'Valid replications: {mask.sum()}/{R_SIMUL}')
print()
print('Test statistic z = beta2_hat / SE(beta2_hat)')
print(f'  Mean z-statistic: {np.mean(z_feedback):.3f}')
print(f'  Median z-statistic: {np.median(z_feedback):.3f}')
print()
print(f'Empirical power (rejection rate when true beta2 = {true_b2})')
print(f'  At 10% level: {power_10:.3f} ({100*power_10:.1f}%)')
print(f'  At 5% level:  {power_05:.3f} ({100*power_05:.1f}%)')
print(f'  At 1% level:  {power_01:.3f} ({100*power_01:.1f}%)')
```

## Bias from Ignoring Feedback: Recursive vs Simultaneous (`tab:mc_bias_comparison`)

Generate data from the simultaneous model but estimate using the recursive ML
estimators (ignoring the feedback parameter $\beta_2$). This demonstrates the
consequence of misspecification when feedback is present.

```python
def run_one_bias_replication(n, p, true_simul, seed):
    """Generate data from simultaneous model, estimate with both FIML and recursive ML."""
    param_names_simul = ['beta0', 'beta1', 'beta2', 'gamma0', 'gamma1', 'gamma2', 'gamma3',
                         'sigma_u_sq', 'sigma_eps_sq']
    param_names_rec_s = ['beta0', 'beta1', 'sigma_u_sq']
    param_names_rec_w = ['gamma0', 'gamma1', 'gamma2', 'gamma3', 'sigma_eps_sq']

    out = {
        'recursive_screen': {k: np.nan for k in param_names_rec_s},
        'recursive_wb': {k: np.nan for k in param_names_rec_w},
        'fiml': {k: np.nan for k in param_names_simul},
    }

    try:
        rng = np.random.default_rng(seed)
        G_sp = generate_network_sparse(n, p, rng)
        I_sp_loc = sp.eye(G_sp.shape[0], format='csr')

        s, y, conv, _ = simulate_simultaneous_data(G_sp, **true_simul, rng=rng)
        if not conv:
            return out

        # Recursive estimation (ignoring feedback)
        res_s = estimate_screen_ml_sparse(s, G_sp)
        for k in param_names_rec_s:
            out['recursive_screen'][k] = res_s[k]

        res_w = estimate_wb_ml_sparse(y, s, G_sp)
        for k in param_names_rec_w:
            out['recursive_wb'][k] = res_w[k]

        # FIML estimation (correct model) -- warm-started from recursive estimates
        ws = build_warm_start(s, y, G_sp)
        res_fiml = estimate_fiml(s, y, G_sp, I_sp_loc, warm_start=ws,
                                 compute_se=False)
        for k in param_names_simul:
            out['fiml'][k] = res_fiml[k]

    except Exception:
        pass

    return out
```

```python
R_BIAS = R_SIMUL
print(f"Running bias comparison MC: R={R_BIAS} (simultaneous data, recursive vs FIML estimation)")
print("This demonstrates the bias from ignoring feedback (beta2).")

seed_seq = np.random.SeedSequence(RNG_SEED + 1000)
child_seqs = seed_seq.spawn(R_BIAS)
bias_seeds = [int(cs.generate_state(1, dtype=np.uint32)[0]) for cs in child_seqs]

t0 = time.time()
with parallel_config(backend='loky', inner_max_num_threads=1):
    bias_results = Parallel(n_jobs=-1)(
        delayed(run_one_bias_replication)(N_SIMUL, P_SIMUL, TRUE_SIMUL, seed)
        for seed in bias_seeds
    )
t_bias = time.time() - t0
print(f"Done in {t_bias:.0f}s ({t_bias/60:.1f} min)")

# Aggregate results
rec_screen = {k: np.array([r['recursive_screen'][k] for r in bias_results]) for k in ['beta0', 'beta1', 'sigma_u_sq']}
rec_wb = {k: np.array([r['recursive_wb'][k] for r in bias_results]) for k in ['gamma0', 'gamma1', 'gamma2', 'gamma3', 'sigma_eps_sq']}
fiml_est = {k: np.array([r['fiml'][k] for r in bias_results]) for k in ['beta0', 'beta1', 'beta2', 'gamma0', 'gamma1', 'gamma2', 'gamma3', 'sigma_u_sq', 'sigma_eps_sq']}
```

```python
# --- Size study results ---

# 1. Wald test size: rejection rate of H0: beta2=0 when beta2=0 is true
est_b2_null = mc_null['estimates']['beta2']
se_b2_null = mc_null['std_errors']['beta2']
mask_null = np.isfinite(est_b2_null) & np.isfinite(se_b2_null) & (se_b2_null > 0)
z_null = est_b2_null[mask_null] / se_b2_null[mask_null]
p_null = 2 * (1 - stats.norm.cdf(np.abs(z_null)))

size_10 = np.mean(p_null < 0.10)
size_05 = np.mean(p_null < 0.05)
size_01 = np.mean(p_null < 0.01)

print('Wald Test Size Under H0: beta2 = 0')
print('=' * 60)
print(f'Valid replications: {mask_null.sum()}/{R_NULL}')
print(f'Mean beta2_hat: {np.mean(est_b2_null[mask_null]):.4f}  (should be near 0)')
print(f'Std(beta2_hat): {np.std(est_b2_null[mask_null], ddof=1):.4f}')
print(f'Mean SE(beta2): {np.mean(se_b2_null[mask_null]):.4f}')
print()
print('Empirical size (rejection rate when H0 is true):')
print(f'  At 10% level: {size_10:.3f}  (nominal: 0.100)')
print(f'  At  5% level: {size_05:.3f}  (nominal: 0.050)')
print(f'  At  1% level: {size_01:.3f}  (nominal: 0.010)')

# MC standard error for size estimate
mc_se_size = np.sqrt(0.05 * 0.95 / mask_null.sum())
print()
print(f'MC standard error for 5% size: {mc_se_size:.4f}')
print(f'95% MC interval for 5% size: [{0.05 - 1.96*mc_se_size:.3f}, {0.05 + 1.96*mc_se_size:.3f}]')

# 2. z-score diagnostics for beta2 under null
print()
print('z-score diagnostics for beta2 under H0:')
print(f'  Mean(z):     {np.mean(z_null):.3f}  (should be near 0)')
print(f'  Std(z):      {np.std(z_null, ddof=1):.3f}  (should be near 1)')
print(f'  Skewness:    {stats.skew(z_null):.3f}')
print(f'  Kurtosis:    {stats.kurtosis(z_null):.3f}')

# 3. Efficiency cost: FIML SE for shared parameters under the null
print()
print('=' * 60)
print('Efficiency: FIML (9 params) estimates under H0 (beta2=0 true)')
print('=' * 60)

shared_params = ['beta1', 'gamma0', 'gamma1', 'gamma2', 'gamma3']
true_null_vals = {k: TRUE_NULL[k] for k in shared_params}

print(f'{"Param":<8} {"True":>6} {"FIML Std(Est.)":>15} {"FIML Mean SE":>13} {"FIML Coverage":>14}')
print('-' * 60)
for k in shared_params:
    true_val = true_null_vals[k]
    est_k = mc_null['estimates'][k]
    se_k = mc_null['std_errors'][k]
    m = np.isfinite(est_k) & np.isfinite(se_k) & (se_k > 0)
    e_vals = est_k[m]
    s_vals = se_k[m]
    std_est = np.std(e_vals, ddof=1)
    mean_se = np.mean(s_vals)
    lower = e_vals - 1.96 * s_vals
    upper = e_vals + 1.96 * s_vals
    cov = np.mean((lower <= true_val) & (true_val <= upper))
    disp = param_names_display.get(k, k)
    print(f'{disp:<8} {true_val:>6.2f} {std_est:>15.4f} {mean_se:>13.4f} {cov:>13.1%}')
```
