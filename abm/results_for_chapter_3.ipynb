## Methods Evaluated
1. **Cross-sectional OLS**: Simple regression of wellbeing on SMU at a single time point
2. **Panel Fixed Effects**: Individual fixed effects to control for time-invariant confounders
3. **Randomized Controlled Trial (RCT)**: Random assignment to treatment (forced SMU reduction)

## True Effect
The true direct effect of SMU on wellbeing is $\beta_1 = -0.5$ (`smu_wb_coeff`).

## Scenarios
1. **Reference**: No COMO, no network effects
2. **COMO**: Population mean positive COMO (no adjacency)
3. **Network and COMO**: Adjacency-weighted COMO on an Erdos--Renyi graph (p = 0.1)

## Key implementation details

- **ABM time stepping.** Each agent updates SMU via an AR(1) process $s_{it} = \rho\, s_{i,t-1} + (1-\rho)\,\mu^s_i + \varepsilon^s_{it}$, then WB is recomputed as $y_{it} = \mu^y_i + \beta_1 s_{it} + \beta_2\,\mathrm{COMO}_{it} + \varepsilon^y_{it}$. The simulation runs for a burn-in period before estimation windows.
- **RCT mechanism.** At a specified treatment time, 20% of agents are randomly assigned to a capped SMU level. The treatment effect is estimated as the mean WB difference between treated and control groups, averaged over post-treatment periods.
- **Network construction.** Erdos--Renyi graphs with connection probability $p = 0.1$; the adjacency matrix is row-normalized so peer averages are unweighted means of neighbours.
- **Parallelisation** via `joblib` with independent seeds per Monte Carlo replication for reproducibility.

## 3. Estimation Methods

We implement three standard estimation approaches used in the SMU-wellbeing literature:

1. **Cross-sectional OLS**: $y_i = \alpha + \beta s_i + \varepsilon_i$
2. **Panel Fixed Effects**: $y_{it} = \alpha_i + \beta s_{it} + \varepsilon_{it}$
3. **RCT**: $\hat{\tau} = \bar{y}^{\text{treated}} - \bar{y}^{\text{control}}$

```python
@dataclass
class EstimationResult:
    """Container for estimation results."""
    method: str
    estimate: float
    se: float | None = None
    n_obs: int = 0
    extra: dict | None = None


def cross_sectional_ols(log: SimulationLog, time_point: int | None = None) -> EstimationResult:
    """
    Cross-sectional OLS: Regress wellbeing on SMU at a single time point.
    
    Model: y_i = alpha + beta * s_i + epsilon_i
    
    Expected failure mode: Confounding from network homophily and correlated effects.
    
    Parameters
    ----------
    log : SimulationLog
        Simulation output containing wellbeing and SMU arrays.
    time_point : int, optional
        Time point to use. If None, uses the final period.
    
    Returns
    -------
    EstimationResult
        Estimated coefficient on SMU.
    """
    T = log.wellbeing.shape[0] - 1
    t = time_point if time_point is not None else T
    
    y = log.wellbeing[t, :]
    x = log.smu[t, :]
    
    # OLS estimation: beta = Cov(x, y) / Var(x)
    n = len(y)
    x_mean = x.mean()
    y_mean = y.mean()
    
    cov_xy = ((x - x_mean) * (y - y_mean)).sum()
    var_x = ((x - x_mean) ** 2).sum()
    
    if var_x < 1e-10:
        return EstimationResult(method="Cross-sectional OLS", estimate=np.nan, n_obs=n)
    
    beta = cov_xy / var_x
    alpha = y_mean - beta * x_mean
    
    # Standard error
    residuals = y - (alpha + beta * x)
    sigma_sq = (residuals ** 2).sum() / (n - 2)
    se_beta = np.sqrt(sigma_sq / var_x)
    
    return EstimationResult(
        method="Cross-sectional OLS",
        estimate=beta,
        se=se_beta,
        n_obs=n,
        extra={"intercept": alpha, "time_point": t}
    )


def panel_fixed_effects(log: SimulationLog, burn_in: int = BURN_IN) -> EstimationResult:
    """
    Panel Fixed Effects: Use within-transformation to eliminate individual fixed effects.
    
    Model: y_it = alpha_i + beta * s_it + epsilon_it
    
    Within-transformation: (y_it - y_i_bar) = beta * (s_it - s_i_bar) + (epsilon_it - epsilon_i_bar)
    
    Expected failure mode: Dynamic feedback violates strict exogeneity.
    
    Parameters
    ----------
    log : SimulationLog
        Simulation output.
    burn_in : int
        Number of initial periods to discard.
    
    Returns
    -------
    EstimationResult
        Estimated coefficient on SMU.
    """
    # Get data after burn-in
    y = log.wellbeing[burn_in:, :]  # (T, n_agents)
    x = log.smu[burn_in:, :]
    
    T, n_agents = y.shape
    
    # Within-transformation: demean by individual
    y_demean = y - y.mean(axis=0, keepdims=True)
    x_demean = x - x.mean(axis=0, keepdims=True)
    
    # Stack into vectors
    y_vec = y_demean.flatten()
    x_vec = x_demean.flatten()
    
    n_obs = len(y_vec)
    
    # OLS on demeaned data
    var_x = (x_vec ** 2).sum()
    if var_x < 1e-10:
        return EstimationResult(method="Panel Fixed Effects", estimate=np.nan, n_obs=n_obs)
    
    cov_xy = (x_vec * y_vec).sum()
    beta = cov_xy / var_x
    
    # Standard error
    residuals = y_vec - beta * x_vec
    sigma_sq = (residuals ** 2).sum() / (n_obs - n_agents - 1)
    se_beta = np.sqrt(sigma_sq / var_x)
    
    return EstimationResult(
        method="Panel Fixed Effects",
        estimate=beta,
        se=se_beta,
        n_obs=n_obs,
        extra={"n_agents": n_agents, "T": T, "burn_in": burn_in}
    )


print(f"Defined Cross-sectional OLS and Panel Fixed Effects methods.")
print(f"RCT method defined separately (requires special simulation).")
```

## 4. Scenario Definitions

We define three scenarios for the Monte Carlo study:

### Scenario 1: Reference
- COMO: None (beta_2 = 0)
- Network: None (no adjacency)
- Delayed effects: None

### Scenario 2: COMO
- **COMO function**: Population mean positive COMO
  - $\text{COMO}_{it} = \max(0, \bar{s}_{t-1} - s_{i,t-1})$
  - Agents feel pressure when using less than the population average
- **Network**: None (agents compare to population mean, not specific neighbors)
- **Delayed effects**: None

### Scenario 3: Network and COMO
- **COMO function**: Adjacency-weighted positive COMO
  - $\text{COMO}_{it} = \max(0, \bar{s}^{\text{neigh}}_{i,t-1} - s_{i,t-1})$
  - Agents compare to their network neighbors
- **Network**: Erdos-Renyi random graph with p = 0.1
- **Delayed effects**: None


```python
def create_reference_scenario(seed: int) -> Scenario:
    """
    Reference scenario where standard estimators should recover beta_1.

    Key features:
    - No COMO (beta_2 = 0)
    - No adjacency (no network)
    - No delayed effects
    - Learning with rate 0.05 (same as other scenarios)
    """
    return Scenario(
        label="Reference",
        seed=seed,
        horizon=HORIZON,
        n_agents=N_AGENTS,
        decision_name="default",
        decision_overrides={
            "rho": RHO,
            "smu_wb_coeff": TRUE_EFFECT,   # beta_1 (target parameter)
            "como_wb_coeff": 0.0,          # beta_2 = 0  => no COMO effect on WB
            "delayed_decline": "False",
            "boost_coeff": 0.0,
            "decline_coeff": 0.0,
            "decline_lag": DECLINE_LAG,    # kept for safety; unused when delayed_decline="False"
            "wellbeing_bounds": WELLBEING_CLIP,
            "smu_bounds": SMU_CLIP,
            "use_hard_clip": True,
        },
        learning_name="regression_based",
        learning_overrides={"learning_rate": LEARNING_RATE},  # same as other scenarios
        como_name="population_mean",                 # computed but irrelevant since como_wb_coeff=0
        como_overrides={},
        adjacency_neighbours=0,
        agent_parameter_overrides=dict(INITIAL_AGENT_PRIORS),
    )


def create_como_scenario(seed: int) -> Scenario:
    """
    COMO scenario: population-mean positive COMO, no adjacency, no delayed effects.
    """
    return Scenario(
        label="COMO",
        seed=seed,
        horizon=HORIZON,
        n_agents=N_AGENTS,
        decision_name="default",
        decision_overrides={
            "rho": RHO,                    # SMU persistence
            "smu_wb_coeff": TRUE_EFFECT,   # Direct SMU -> WB effect (TRUE EFFECT)
            "como_wb_coeff": COMO_COEFF,   # COMO -> WB effect
            "delayed_decline": "False",
            "boost_coeff": 0.0,
            "decline_coeff": 0.0,
            "decline_lag": DECLINE_LAG,
            "wellbeing_bounds": WELLBEING_CLIP,
            "smu_bounds": SMU_CLIP,
            "use_hard_clip": True,
        },
        learning_name="regression_based",
        learning_overrides={"learning_rate": LEARNING_RATE},
        como_name="population_mean_positive",
        como_overrides={},
        adjacency_neighbours=0,  # No explicit network
        agent_parameter_overrides=dict(INITIAL_AGENT_PRIORS),
    )


def create_network_como_scenario(seed: int, p: float = 0.1) -> Scenario:
    """
    Network and COMO scenario: adjacency-weighted COMO with no delayed effects.

    Network structure: Erdos-Renyi random graph with p = 0.1.
    """
    return Scenario(
        label="Network and COMO",
        seed=seed,
        horizon=HORIZON,
        n_agents=N_AGENTS,
        decision_name="default",
        decision_overrides={
            "rho": RHO,
            "smu_wb_coeff": TRUE_EFFECT,
            "como_wb_coeff": COMO_COEFF,
            "delayed_decline": "False",
            "boost_coeff": 0.0,
            "decline_coeff": 0.0,
            "decline_lag": DECLINE_LAG,
            "wellbeing_bounds": WELLBEING_CLIP,
            "smu_bounds": SMU_CLIP,
            "use_hard_clip": True,
        },
        learning_name="regression_based",
        learning_overrides={"learning_rate": LEARNING_RATE},
        como_name="adjacency_positive",
        como_overrides={"adjacency_kind": "erdos_renyi", "p": float(p), "undirected": True},
        adjacency_neighbours=0,
        agent_parameter_overrides=dict(INITIAL_AGENT_PRIORS),
    )


# Define scenario factories
SCENARIO_FACTORIES = {
    "Reference": create_reference_scenario,
    "COMO": create_como_scenario,
    "Network and COMO": lambda seed: create_network_como_scenario(seed, p=0.1),
}

print(f"Defined {len(SCENARIO_FACTORIES)} scenario types for main study:")
for name in SCENARIO_FACTORIES:
    print(f"  - {name}")
```

## 5. Simulation Infrastructure

Helper functions to run simulations and collect estimates.

```python
def build_erdos_renyi_adjacency(
    n_agents: int,
    p: float,
    *,
    rng: np.random.Generator,
    undirected: bool = True,
) -> np.ndarray:
    """Create an Erdos-Renyi adjacency matrix with edge probability p."""
    p = float(p)
    if p < 0.0 or p > 1.0:
        raise ValueError("p must be between 0 and 1")

    if undirected:
        draws = rng.random((n_agents, n_agents))
        upper = np.triu(np.ones((n_agents, n_agents), dtype=bool), k=1)
        mask = upper & (draws < p)
        adjacency = np.zeros((n_agents, n_agents), dtype=float)
        adjacency[mask] = 1.0
        adjacency = adjacency + adjacency.T
    else:
        adjacency = (rng.random((n_agents, n_agents)) < p).astype(float)
        np.fill_diagonal(adjacency, 0.0)

    # Connect any isolated nodes to one randomly chosen other agent
    for i in range(n_agents):
        if adjacency[i].sum() == 0.0:
            others = [j for j in range(n_agents) if j != i]
            j = rng.choice(others)
            adjacency[i, j] = 1.0
            if undirected:
                adjacency[j, i] = 1.0

    return adjacency


def _initialise_agent_parameters(
    rng: np.random.Generator,
    overrides: dict[str, float],
) -> tuple[float, float, dict[str, float]]:
    """Initialize agent parameters with random draws from priors."""
    override_params = {str(key): float(value) for key, value in overrides.items()}

    # Draw from priors (matching Table 3.1)
    wb_mu = float(np.clip(rng.normal(override_params.get("wb_mu", 6.0), 1.0), 3.0, 9.0))
    wb_sigma = max(float(np.clip(rng.normal(override_params.get("wb_sigma", 1.0), 0.1), 0.1, 1.5)), 1e-6)
    smu_mu = float(np.clip(rng.normal(override_params.get("smu_mu", 3.0), 0.2), 1.0, 8.0))
    smu_sigma = max(float(np.clip(rng.normal(override_params.get("smu_sigma", 1.0), 0.1), 0.1, 1.5)), 1e-6)

    initial_wb = float(np.clip(rng.normal(wb_mu, wb_sigma), WELLBEING_CLIP[0], WELLBEING_CLIP[1]))
    initial_smu = float(np.clip(rng.normal(smu_mu, smu_sigma), SMU_CLIP[0], SMU_CLIP[1]))

    parameters = {
        "wb_mu": wb_mu,
        "wb_sigma": wb_sigma,
        "smu_mu": smu_mu,
        "smu_sigma": smu_sigma,
        "wb_baseline": initial_wb,
        "smu_baseline": initial_smu,
    }

    if "learning_rate" in override_params:
        parameters["learning_rate"] = float(override_params["learning_rate"])

    return initial_wb, initial_smu, parameters


def run_single_simulation(scenario: Scenario) -> SimulationLog:
    """
    Run a single ABM simulation for the given scenario.

    Returns
    -------
    SimulationLog
        Contains wellbeing, smu, como arrays of shape (T+1, n_agents)
    """
    rng = np.random.default_rng(scenario.seed)
    adjacency_rng = np.random.default_rng(scenario.seed + 991)

    # Build decision rule
    decision_rule = default_decision(**scenario.decision_overrides)

    # Build learning rule
    learning_rule = regression_based_learning(**scenario.learning_overrides)

    # Create agents
    agents = []
    for idx in range(scenario.n_agents):
        initial_wb, initial_smu, parameters = _initialise_agent_parameters(
            rng, scenario.agent_parameter_overrides
        )
        config = AgentConfig(
            initial_wellbeing=initial_wb,
            initial_smu=initial_smu,
            parameters=parameters,
            label=f"Agent {idx}",
        )
        agents.append(
            BaseAgent(
                config=config,
                decision_rule=decision_rule,
                learning_rule=learning_rule,
            )
        )

    # Resolve COMO function and adjacency
    adjacency_matrix = None
    if scenario.como_name == "population_mean":
        como_function = population_mean_como
    elif scenario.como_name == "population_mean_positive":
        como_function = population_mean_positive_como
    elif scenario.como_name in ("adjacency", "adjacency_positive"):
        adjacency_kind = str(scenario.como_overrides.get("adjacency_kind", "ring"))
        if adjacency_kind == "erdos_renyi":
            p = float(scenario.como_overrides.get("p", 0.1))
            undirected = bool(scenario.como_overrides.get("undirected", True))
            adjacency_matrix = build_erdos_renyi_adjacency(
                scenario.n_agents, p, rng=adjacency_rng, undirected=undirected
            )
        else:
            adjacency_matrix = build_ring_adjacency(scenario.n_agents, scenario.adjacency_neighbours)
        if scenario.como_name == "adjacency":
            como_function = adjacency_weighted_como
        else:
            como_function = adjacency_weighted_positive_como
    else:
        como_function = population_mean_como

    # Set agent indices and neighbour weights
    for idx, agent in enumerate(agents):
        setattr(agent, "agent_index", idx)
        setattr(agent, "neighbour_weights", [])

    if adjacency_matrix is not None:
        weights = np.array(adjacency_matrix, dtype=float)
        row_sums = weights.sum(axis=1, keepdims=True)
        with np.errstate(divide="ignore", invalid="ignore"):
            weights = np.where(row_sums > 0, weights / row_sums, 0.0)
        for idx, agent in enumerate(agents):
            neighbours = [
                (int(neighbour_idx), float(weights[idx, neighbour_idx]))
                for neighbour_idx in range(weights.shape[1])
                if weights[idx, neighbour_idx] > 0 and neighbour_idx != idx
            ]
            setattr(agent, "neighbour_weights", neighbours)

    # Run simulation
    runner = ABDRunner(
        agents=agents,
        T=scenario.horizon,
        como_function=como_function,
        adjacency_matrix=adjacency_matrix,
        seed=scenario.seed,
    )
    log = runner.run()
    log.metadata["scenario_label"] = scenario.label
    log.metadata["seed"] = scenario.seed

    return log


print("Simulation infrastructure ready.")
```

## 6. RCT Implementation

The RCT method requires a modified simulation where:
1. Run simulation until treatment time
2. Randomly assign 20% of agents to treatment (hard SMU cap of 1 hour/day)
3. Continue simulation with treated agents having their SMU capped at maximum 1 hour
4. Compare mean wellbeing between treatment and control groups

**Expected failure mode**: SUTVA violation. When treated agents reduce their SMU, the reference level for untreated agents changes, altering their COMO. The control group is thus indirectly "treated" through the network.

```python
def make_capped_decision_rule(
    original_rule: Callable,
    smu_cap: float,
    treatment_start: int,
    smu_wb_coeff: float,
    boost_coeff: float,
    wellbeing_bounds: tuple[float, float] = (0.0, 10.0),
) -> Callable:
    """
    Wrap a decision rule to enforce an SMU cap after treatment starts.

    This ensures the SMU cap is applied and wellbeing is properly recomputed
    using the unclipped wellbeing value to avoid bias from clipping.
    """
    def capped_rule(agent, context):
        # Call original decision rule
        result = original_rule(agent, context)

        # Check if this agent is treated and we're past treatment start
        is_treated = agent.parameters.get("treated", 0.0) > 0.5
        t = context.time_index

        if is_treated and t >= treatment_start:
            # Cap the SMU
            capped_smu = min(result.smu, smu_cap)

            # Recompute wellbeing with the capped SMU if SMU changed
            if capped_smu < result.smu:
                # IMPORTANT: Use unclipped wellbeing to avoid bias from clipping
                wellbeing_unclipped = result.logs.get("wellbeing_unclipped", result.wellbeing)

                # Adjust wellbeing for the SMU reduction
                smu_diff = capped_smu - result.smu
                adjusted_wb = wellbeing_unclipped + smu_wb_coeff * smu_diff
                if boost_coeff != 0.0:
                    adjusted_wb += boost_coeff * smu_diff

                # Apply hard clipping to the adjusted value
                adjusted_wb = float(np.clip(adjusted_wb, wellbeing_bounds[0], wellbeing_bounds[1]))

                return AgentStepResult(
                    wellbeing=adjusted_wb,
                    smu=capped_smu,
                    logs=result.logs,
                )

        return result

    return capped_rule


def _compute_rct_effect(
    log: SimulationLog,
    treated_mask: np.ndarray,
    treatment_start: int,
) -> tuple[float, int, dict[str, float]]:
    """Compute the scaled RCT effect using the final time step (single endpoint).
    
    This matches how real RCTs measure outcomes: at a single assessment point
    after the treatment period, rather than averaging over all post-treatment
    time steps. Using a single endpoint makes the RCT's effective sample size
    equal to the number of agents (n_treated + n_control), directly comparable
    to the cross-sectional OLS which also uses a single time point.
    """
    final_wb = log.wellbeing[-1, :]  # shape: (n_agents,)
    final_smu = log.smu[-1, :]       # shape: (n_agents,)

    n_obs = treated_mask.size
    if final_wb.size == 0:
        return float("nan"), 0, {}

    mean_wb_treated = float(final_wb[treated_mask].mean())
    mean_wb_control = float(final_wb[~treated_mask].mean())
    mean_smu_treated = float(final_smu[treated_mask].mean())
    mean_smu_control = float(final_smu[~treated_mask].mean())

    denom = mean_smu_treated - mean_smu_control
    if abs(denom) < 1e-10:
        return float("nan"), n_obs, {}

    effect = (mean_wb_treated - mean_wb_control) / denom
    stats = {
        "mean_wb_treated": mean_wb_treated,
        "mean_wb_control": mean_wb_control,
        "mean_smu_treated": mean_smu_treated,
        "mean_smu_control": mean_smu_control,
    }
    return float(effect), n_obs, stats


def run_rct_simulation(scenario: Scenario) -> tuple[SimulationLog, np.ndarray, float]:
    """
    Run an RCT simulation with a hard SMU cap for treated agents.

    Returns
    -------
    SimulationLog, treated_mask, rct_effect
    """
    rng = np.random.default_rng(scenario.seed)
    adjacency_rng = np.random.default_rng(scenario.seed + 991)

    # Use a separate RNG for treatment assignment so that agent initialisation

    # draws match run_single_simulation for the same seed.
    treatment_rng = np.random.default_rng(scenario.seed + 777)
    treated_mask = treatment_rng.random(scenario.n_agents) < RCT_TREATMENT_FRACTION

    decision_overrides = dict(scenario.decision_overrides)
    decision_rule = default_decision(**decision_overrides)
    capped_rule = make_capped_decision_rule(
        decision_rule,
        smu_cap=RCT_SMU_CAP,
        treatment_start=RCT_TREATMENT_START,
        smu_wb_coeff=decision_overrides.get("smu_wb_coeff", TRUE_EFFECT),
        boost_coeff=decision_overrides.get("boost_coeff", 0.0),
        wellbeing_bounds=decision_overrides.get("wellbeing_bounds", WELLBEING_CLIP),
    )

    learning_rule = regression_based_learning(**scenario.learning_overrides)

    agents = []
    for idx in range(scenario.n_agents):
        initial_wb, initial_smu, parameters = _initialise_agent_parameters(
            rng, scenario.agent_parameter_overrides
        )
        parameters["treated"] = float(treated_mask[idx])
        config = AgentConfig(
            initial_wellbeing=initial_wb,
            initial_smu=initial_smu,
            parameters=parameters,
            label=f"Agent {idx}",
        )
        agents.append(
            BaseAgent(
                config=config,
                decision_rule=capped_rule,
                learning_rule=learning_rule,
            )
        )

    adjacency_matrix = None
    if scenario.como_name == "population_mean":
        como_function = population_mean_como
    elif scenario.como_name == "population_mean_positive":
        como_function = population_mean_positive_como
    elif scenario.como_name in ("adjacency", "adjacency_positive"):
        adjacency_kind = str(scenario.como_overrides.get("adjacency_kind", "ring"))
        if adjacency_kind == "erdos_renyi":
            p = float(scenario.como_overrides.get("p", 0.1))
            undirected = bool(scenario.como_overrides.get("undirected", True))
            adjacency_matrix = build_erdos_renyi_adjacency(
                scenario.n_agents, p, rng=adjacency_rng, undirected=undirected
            )
        else:
            adjacency_matrix = build_ring_adjacency(scenario.n_agents, scenario.adjacency_neighbours)
        if scenario.como_name == "adjacency":
            como_function = adjacency_weighted_como
        else:
            como_function = adjacency_weighted_positive_como
    else:
        como_function = population_mean_como

    for idx, agent in enumerate(agents):
        setattr(agent, "agent_index", idx)
        setattr(agent, "neighbour_weights", [])

    if adjacency_matrix is not None:
        weights = np.array(adjacency_matrix, dtype=float)
        row_sums = weights.sum(axis=1, keepdims=True)
        with np.errstate(divide="ignore", invalid="ignore"):
            weights = np.where(row_sums > 0, weights / row_sums, 0.0)
        for idx, agent in enumerate(agents):
            neighbours = [
                (int(neighbour_idx), float(weights[idx, neighbour_idx]))
                for neighbour_idx in range(weights.shape[1])
                if weights[idx, neighbour_idx] > 0 and neighbour_idx != idx
            ]
            setattr(agent, "neighbour_weights", neighbours)

    runner = ABDRunner(
        agents=agents,
        T=scenario.horizon,
        como_function=como_function,
        adjacency_matrix=adjacency_matrix,
        seed=scenario.seed,
    )
    log = runner.run()
    log.metadata["scenario_label"] = scenario.label
    log.metadata["seed"] = scenario.seed

    rct_effect, _, _ = _compute_rct_effect(log, treated_mask, RCT_TREATMENT_START)

    return log, treated_mask, rct_effect


def rct_estimator(
    scenario_factory: Callable[[int], Scenario],
    seed: int,
    treatment_start: int = RCT_TREATMENT_START,
) -> EstimationResult:
    """Compute the scaled RCT estimate for a scenario."""
    scenario = scenario_factory(seed)
    log, treated_mask, _ = run_rct_simulation(scenario)

    effect, n_obs, stats = _compute_rct_effect(log, treated_mask, treatment_start)

    return EstimationResult(
        method="RCT",
        estimate=effect,
        se=None,
        n_obs=n_obs,
        extra=stats,
    )

print("Capped decision rule helper and RCT utilities defined.")
```

## 8. Monte Carlo Study

Run multiple simulations per scenario and collect estimation results for all three methods.

```python
def run_monte_carlo_study(
    scenario_factory: Callable[[int], Scenario],
    n_runs: int,
    base_seed: int = 42,
    burn_in: int = BURN_IN,
    n_jobs: int | None = None,
) -> pd.DataFrame:
    """
    Run Monte Carlo study for a given scenario.

    For each run:
    1. Run standard simulation and apply Cross-sectional OLS and Panel FE
    2. Run RCT simulation and compute treatment effect

    Returns DataFrame with columns: run, method, estimate, se, bias
    """
    import os

    try:
        from joblib import Parallel, delayed
        has_joblib = True
    except ImportError:
        has_joblib = False

    def _run_one(run_idx: int) -> list[dict[str, object]]:
        seed = base_seed + run_idx
        scenario = scenario_factory(seed)

        run_rows: list[dict[str, object]] = []

        # Run standard simulation for OLS and FE
        log = run_single_simulation(scenario)

        # Cross-sectional OLS
        ols_result = cross_sectional_ols(log)
        run_rows.append({
            "run": run_idx,
            "seed": seed,
            "method": ols_result.method,
            "estimate": ols_result.estimate,
            "se": ols_result.se,
            "n_obs": ols_result.n_obs,
            "bias": ols_result.estimate - TRUE_EFFECT,
        })

        # Panel Fixed Effects
        fe_result = panel_fixed_effects(log, burn_in=burn_in)
        run_rows.append({
            "run": run_idx,
            "seed": seed,
            "method": fe_result.method,
            "estimate": fe_result.estimate,
            "se": fe_result.se,
            "n_obs": fe_result.n_obs,
            "bias": fe_result.estimate - TRUE_EFFECT,
        })

        # RCT
        # Create a large-n version for RCT
        def rct_scenario_large(s):
            sc = scenario_factory(s)
            # RCT needs more agents to compensate for the treatment/control split
            sc.n_agents = 500  # Override n_agents for RCT
            return sc

        rct_result = rct_estimator(rct_scenario_large, seed)

        run_rows.append({
            "run": run_idx,
            "seed": seed,
            "method": rct_result.method,
            "estimate": rct_result.estimate,
            "se": rct_result.se,
            "n_obs": rct_result.n_obs,
            "bias": rct_result.estimate - TRUE_EFFECT if not np.isnan(rct_result.estimate) else np.nan,
        })

        return run_rows

    if n_runs <= 0:
        return pd.DataFrame()

    max_workers = n_jobs if n_jobs is not None else (os.cpu_count() or 1)
    max_workers = max(1, min(int(max_workers), int(n_runs)))

    results: list[dict[str, object]] = []

    if max_workers == 1:
        for run_idx in tqdm(range(n_runs), desc="Monte Carlo runs"):
            results.extend(_run_one(run_idx))
    elif has_joblib:
        mapped = Parallel(n_jobs=max_workers, backend="loky")(
            delayed(_run_one)(run_idx)
            for run_idx in tqdm(range(n_runs), desc="Monte Carlo runs")
        )
        for run_rows in mapped:
            results.extend(run_rows)
    else:
        from concurrent.futures import ThreadPoolExecutor

        print("joblib not installed; falling back to thread-based parallelism.")
        with ThreadPoolExecutor(max_workers=max_workers) as executor:
            mapped = executor.map(_run_one, range(n_runs))
            for run_rows in tqdm(mapped, total=n_runs, desc="Monte Carlo runs"):
                results.extend(run_rows)

    return pd.DataFrame(results)


def compute_summary_statistics(results_df: pd.DataFrame) -> pd.DataFrame:
    """Compute summary statistics for Monte Carlo results."""
    summary = results_df.groupby(["scenario", "method"]).agg(
        mean_estimate=("estimate", "mean"),
        std_estimate=("estimate", "std"),
        mean_bias=("bias", "mean"),
        std_bias=("bias", "std"),
        rmse=("bias", lambda x: np.sqrt((x ** 2).mean())),
        median_estimate=("estimate", "median"),
        q05=("estimate", lambda x: x.quantile(0.05)),
        q95=("estimate", lambda x: x.quantile(0.95)),
        n_valid=("estimate", lambda x: x.notna().sum()),
    ).reset_index()

    summary["true_effect"] = TRUE_EFFECT
    summary["bias_pct"] = (summary["mean_bias"] / abs(TRUE_EFFECT)) * 100

    return summary


print("Monte Carlo study functions ready.")
```

```python
# Run Monte Carlo for all scenarios
print("=" * 70)
print(f"MONTE CARLO STUDY: {N_MC_RUNS} runs per scenario")
print("=" * 70)

scenario_order = ["Reference", "COMO", "Network and COMO"]
base_seeds = {"Reference": 1000, "COMO": 2000, "Network and COMO": 3000}
all_scenario_results = []

for name in scenario_order:
    print(f"\n--- {name} ---")
    scenario_results = run_monte_carlo_study(
        scenario_factory=SCENARIO_FACTORIES[name],
        n_runs=N_MC_RUNS,
        base_seed=base_seeds[name],
    )
    scenario_results["scenario"] = name
    all_scenario_results.append(scenario_results)
    print(f"Completed. Total estimates: {len(scenario_results)}")

# Combine results
all_results = pd.concat(all_scenario_results, ignore_index=True)
print(f"\nTotal results: {len(all_results)}")

# Reference Monte Carlo summary (sanity check)
reference_results = all_results[all_results["scenario"] == "Reference"].copy()
reference_summary = compute_summary_statistics(reference_results)

print("\nReference Monte Carlo summary:")
display_cols = ["scenario", "method", "mean_estimate", "mean_bias", "bias_pct"]
reference_display = reference_summary[display_cols].copy()
reference_display.columns = ["Scenario", "Method", "Mean Estimate", "Bias", "Bias (%)"]
reference_display["Mean Estimate"] = reference_display["Mean Estimate"].map(lambda x: f"{x:.3f}")
reference_display["Bias"] = reference_display["Bias"].map(lambda x: f"{x:.3f}")
reference_display["Bias (%)"] = reference_display["Bias (%)"].map(lambda x: f"{x:.1f}%")
print(reference_display.to_string(index=False))

# Check reference recovery (tolerance relaxed for RCT with hard cap intervention)
tol_ols_fe = 0.15  # Stricter tolerance for OLS and FE
tol_rct = 0.30     # Looser tolerance for RCT (hard cap creates more variance)
mean_by_method = reference_summary.set_index("method")["mean_estimate"].to_dict()
for m in ["Cross-sectional OLS", "Panel Fixed Effects", "RCT"]:
    if m in mean_by_method and np.isfinite(mean_by_method[m]):
        err = abs(mean_by_method[m] - TRUE_EFFECT)
        tol = tol_rct if m == "RCT" else tol_ols_fe
        print(f"\n|Mean({m}) - TRUE_EFFECT| = {err:.3f} (tolerance: {tol:.2f})")
        assert err < tol, f"Reference recovery failed for {m}: error {err:.3f} >= {tol:.2f}"

print("\nOK: Reference sanity check passed (all methods recover beta_1 within tolerance).")
```

```python
def plot_forest_style(summary_df: pd.DataFrame, figsize=(10, 7)):
    """Create a publication-quality forest-plot style visualization of mean estimates with CI."""
    
    # Set publication-quality style
    plt.rcParams.update({
        'font.size': 11,
        'axes.titlesize': 13,
        'axes.labelsize': 11,
        'xtick.labelsize': 10,
        'ytick.labelsize': 10,
        'legend.fontsize': 10,
        'font.family': 'serif',
        'axes.spines.top': False,
        'axes.spines.right': False,
    })
    
    fig, ax = plt.subplots(figsize=figsize)

    # Define method order and scenario order for logical grouping
    method_order = ["Cross-sectional OLS", "Panel Fixed Effects", "RCT"]
    scenario_order = ["Reference", "COMO", "Network and COMO"]
    
    # Sort for consistent ordering: group by scenario, then by method
    summary_df = summary_df.copy()
    summary_df["method_rank"] = summary_df["method"].map({m: i for i, m in enumerate(method_order)})
    summary_df["scenario_rank"] = summary_df["scenario"].map({s: i for i, s in enumerate(scenario_order)})
    summary_df = summary_df.sort_values(["scenario_rank", "method_rank"], ascending=[False, False]).reset_index(drop=True)
    
    summary_df["label"] = summary_df["method"] + "\n(" + summary_df["scenario"] + ")"

    y_positions = range(len(summary_df))
    
    # Professional color palette
    color_map = {
        "Reference": "#7F7F7F",       # Gray
        "COMO": "#4878A8",            # Steel blue
        "Network and COMO": "#DD8452", # Muted orange
    }
    colors = [color_map.get(s, "#7F7F7F") for s in summary_df["scenario"]]

    # Calculate 95% CI of the mean
    ci = 1.96 * summary_df["std_estimate"] / np.sqrt(N_MC_RUNS)
    
    # Plot error bars
    ax.errorbar(
        summary_df["mean_estimate"],
        y_positions,
        xerr=ci,
        fmt="none",
        capsize=4,
        capthick=1.5,
        elinewidth=1.5,
        color="#333333",
        zorder=3,
    )

    # Plot markers with scenario colors
    for i, (est, color) in enumerate(zip(summary_df["mean_estimate"], colors)):
        ax.scatter([est], [i], color=color, s=120, zorder=5, edgecolor='white', linewidth=1.5)

    # True effect line
    ax.axvline(TRUE_EFFECT, color="#C44E52", linestyle="--", linewidth=2, 
               label=f"True effect ({TRUE_EFFECT})", zorder=2)
    
    # Zero reference line
    ax.axvline(0, color="#AAAAAA", linestyle=":", linewidth=1, alpha=0.7, zorder=1)

    ax.set_yticks(y_positions)
    ax.set_yticklabels(summary_df["label"])
    ax.set_xlabel("Estimated Effect (with 95% CI)")
    ax.set_title(f"Mean Estimates Across Methods and Scenarios\n(M = {N_MC_RUNS} Monte Carlo runs)",
                 fontweight="bold", pad=15)
    
    # Add subtle horizontal grid
    ax.xaxis.grid(True, linestyle='-', alpha=0.3, color='gray', zorder=0)
    ax.set_axisbelow(True)

    # Add bias annotations with improved positioning
    for i, (est, bias, ci_val) in enumerate(zip(summary_df["mean_estimate"], summary_df["mean_bias"], ci)):
        # Position text to the right of the CI
        text_x = est + ci_val + 0.02
        ax.annotate(
            f"Bias: {bias:+.3f}",
            xy=(text_x, i),
            fontsize=9,
            va="center",
            ha="left",
            color="#333333",
        )

    # Create custom legend for scenarios
    from matplotlib.lines import Line2D
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', markerfacecolor=color_map["Reference"], 
               markersize=10, label='Reference'),
        Line2D([0], [0], marker='o', color='w', markerfacecolor=color_map["COMO"], 
               markersize=10, label='COMO'),
        Line2D([0], [0], marker='o', color='w', markerfacecolor=color_map["Network and COMO"], 
               markersize=10, label='Network and COMO'),
        Line2D([0], [0], color='#C44E52', linestyle='--', linewidth=2, label=f'True effect ({TRUE_EFFECT})'),
    ]
    ax.legend(handles=legend_elements, loc="upper right", framealpha=0.95, edgecolor='none')

    # Adjust x-axis limits to accommodate bias labels
    x_min, x_max = ax.get_xlim()
    ax.set_xlim(x_min, x_max + 0.15)

    plt.tight_layout()
    return fig


fig_forest = plot_forest_style(summary_stats)
fig_forest.savefig(os.path.join(OUTPUT_DIR, 'abm_forest.pdf'), dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("Figure saved: abm_forest.pdf")
```

## 11. Delayed Decline Kernel Figure

Visualize the delayed decline kernel used in the Simpson's Paradox scenario (Section 3.7).

```python
def plot_delayed_decline_kernel(output_dir, figsize=(12, 4.5)):
    """Create delayed-decline kernel figure with components and cumulative effect."""
    plt.rcParams.update({
        'font.size': 11, 'axes.titlesize': 13, 'axes.labelsize': 11,
        'xtick.labelsize': 10, 'ytick.labelsize': 10, 'legend.fontsize': 10,
        'font.family': 'serif', 'axes.spines.top': False, 'axes.spines.right': False,
    })

    # Parameters from the Simpson's Paradox scenario
    REWARD_SCALE = 0.5   # alpha_r
    PENALTY_SCALE = 0.05  # alpha_p
    FAST_TIMESCALE = 1.0  # tau_f
    SLOW_TIMESCALE = 15.0  # tau_s
    EFFECT_HORIZON = 15   # H

    COLOR_REWARD = "#2E8B57"
    COLOR_PENALTY = "#C44E52"
    COLOR_COMBINED = "#4878A8"
    COLOR_ANNOTATION = "#333333"

    t = np.arange(EFFECT_HORIZON, dtype=float)
    reward_component = REWARD_SCALE * np.exp(-t / FAST_TIMESCALE)
    penalty_component = PENALTY_SCALE * np.exp(-t / SLOW_TIMESCALE)
    kernel = reward_component - penalty_component
    cumulative = np.cumsum(kernel)

    fig, axes = plt.subplots(1, 2, figsize=figsize)

    ax1 = axes[0]
    ax1.plot(t, reward_component, color=COLOR_REWARD, linewidth=2,
             label=r"Reward ($\alpha_r e^{-t/\tau_f}$)")
    ax1.plot(t, penalty_component, color=COLOR_PENALTY, linewidth=2,
             label=r"Penalty ($\alpha_p e^{-t/\tau_s}$)")
    ax1.plot(t, kernel, color=COLOR_COMBINED, linewidth=2.5, label="Combined kernel")
    ax1.axhline(y=0, color="#AAAAAA", linestyle="-", linewidth=0.8, alpha=0.7)
    ax1.fill_between(t, 0, kernel, where=kernel >= 0, alpha=0.25, color=COLOR_REWARD)
    ax1.fill_between(t, 0, kernel, where=kernel < 0, alpha=0.25, color=COLOR_PENALTY)
    ax1.set_xlabel(r"Days since SMU ($\tau$)")
    ax1.set_ylabel("Effect on WB")
    ax1.set_title("Delayed Decline Kernel Components", fontweight="medium")
    ax1.legend(loc="upper right", framealpha=0.95, edgecolor="none")
    ax1.set_xlim(0, EFFECT_HORIZON - 1)
    ax1.yaxis.grid(True, linestyle="-", alpha=0.3, color="gray")
    ax1.set_axisbelow(True)

    param_text = (
        f'$\\alpha_r = {REWARD_SCALE}$\n'
        f'$\\alpha_p = {PENALTY_SCALE}$\n'
        f'$\\tau_f = {FAST_TIMESCALE}$\n'
        f'$\\tau_s = {SLOW_TIMESCALE}$\n'
        f'$H = {EFFECT_HORIZON}$'
    )
    ax1.text(0.65, 0.55, param_text, transform=ax1.transAxes, fontsize=10,
             verticalalignment="top", color=COLOR_ANNOTATION,
             bbox=dict(boxstyle="round", facecolor="white", alpha=0.9, edgecolor="none"))

    ax2 = axes[1]
    ax2.plot(t, cumulative, color=COLOR_COMBINED, linewidth=2.5)
    ax2.axhline(y=0, color="#AAAAAA", linestyle="-", linewidth=0.8, alpha=0.7)
    ax2.fill_between(t, 0, cumulative, where=cumulative >= 0, alpha=0.25, color=COLOR_REWARD)
    ax2.fill_between(t, 0, cumulative, where=cumulative < 0, alpha=0.25, color=COLOR_PENALTY)

    zero_crossings = np.where(np.diff(np.sign(cumulative)))[0]
    for zc in zero_crossings:
        ax2.axvline(x=zc + 0.5, color="#7F7F7F", linestyle="--", linewidth=1.2, alpha=0.8)
        ax2.annotate(f"~{zc + 1} days", xy=(zc + 0.5, 0),
                     xytext=(zc + 1.5, cumulative.max() * 0.3),
                     fontsize=9, color=COLOR_ANNOTATION,
                     arrowprops=dict(arrowstyle="->", color="#7F7F7F", alpha=0.8))

    ax2.set_xlabel(r"Days since SMU ($\tau$)")
    ax2.set_ylabel("Cumulative effect on WB")
    ax2.set_title("Cumulative Effect Over Time", fontweight="medium")
    ax2.set_xlim(0, EFFECT_HORIZON - 1)
    ax2.yaxis.grid(True, linestyle="-", alpha=0.3, color="gray")
    ax2.set_axisbelow(True)

    total_effect = cumulative[-1]
    ax2.annotate(f"Total effect: {total_effect:.2f}", xy=(EFFECT_HORIZON - 1, total_effect),
                 xytext=(EFFECT_HORIZON - 5, total_effect - 0.3), fontsize=10,
                 color=COLOR_ANNOTATION,
                 arrowprops=dict(arrowstyle="->", color=COLOR_ANNOTATION, alpha=0.8))

    plt.tight_layout()
    save_path = os.path.join(output_dir, "delayed_decline_kernel.pdf")
    fig.savefig(save_path, bbox_inches="tight", facecolor="white")
    print(f"Saved: {save_path}")
    print(f"Kernel statistics:")
    print(f"  Immediate reward (t=0): {kernel[0]:.3f}")
    print(f"  Total effect (sum): {kernel.sum():.3f}")
    print(f"  Max positive: {kernel.max():.3f} at t={kernel.argmax()}")
    print(f"  Min (max penalty): {kernel.min():.3f} at t={kernel.argmin()}")
    plt.show()
    return fig


fig_kernel = plot_delayed_decline_kernel(OUTPUT_DIR)
```

```python
# Plotting helpers for Simpson's Paradox figures
# (Professional thesis-style matching other chapters)

THESIS_COLORS = {
    "steel_blue": "#4878A8",
    "muted_orange": "#DD8452",
    "muted_red": "#C44E52",
    "gray": "#7F7F7F",
    "dark_gray": "#333333",
    "light_gray": "#AAAAAA",
    "sea_green": "#2E8B57",
}


def plot_simpsons_population_means(log, output_dir, figsize=(10, 5)):
    """Plot population means for wellbeing and SMU over time."""
    plt.rcParams.update({
        'font.size': 11, 'axes.titlesize': 13, 'axes.labelsize': 11,
        'xtick.labelsize': 10, 'ytick.labelsize': 10, 'legend.fontsize': 10,
        'font.family': 'serif', 'axes.spines.top': False, 'axes.spines.right': False,
    })

    time_index = log.metadata.get("time_index", np.arange(log.wellbeing.shape[0]))

    fig, ax = plt.subplots(figsize=figsize)
    ax.plot(time_index, log.wellbeing.mean(axis=1), label="Mean Well-being",
            color=THESIS_COLORS["steel_blue"], linewidth=2)
    ax.plot(time_index, log.smu.mean(axis=1), label="Mean SMU",
            color=THESIS_COLORS["muted_orange"], linewidth=2)

    ax.set_xlabel("Time (days)")
    ax.set_ylabel("Population Mean")
    ax.set_title("Population Averages Over Time", fontweight="medium")
    ax.legend(loc="best", framealpha=0.95, edgecolor="none")
    ax.yaxis.grid(True, linestyle="-", alpha=0.3, color="gray")
    ax.set_axisbelow(True)

    fig.tight_layout()

    save_path = os.path.join(output_dir, "abm_simpsons_trends.pdf")
    fig.savefig(save_path, dpi=300, bbox_inches="tight", facecolor="white")
    print(f"Saved: {save_path}")
    plt.show()
    return fig


def plot_simpsons_regression_overlay(log, output_dir, figsize=(14, 8)):
    """Plot mean wellbeing with inset cross-sectional regressions at chosen times."""
    plt.rcParams.update({
        'font.size': 11, 'axes.titlesize': 13, 'axes.labelsize': 11,
        'xtick.labelsize': 10, 'ytick.labelsize': 10, 'legend.fontsize': 10,
        'font.family': 'serif', 'axes.spines.top': False, 'axes.spines.right': False,
    })

    n_steps = log.wellbeing.shape[0]
    time_index = log.metadata.get("time_index", np.arange(n_steps))
    time_index = np.asarray(time_index)
    mean_wb = log.wellbeing.mean(axis=1)

    fig, ax_main = plt.subplots(figsize=figsize)
    ax_main.plot(time_index, mean_wb, color=THESIS_COLORS["steel_blue"],
                 linewidth=2, label="Mean Well-being")
    ax_main.set_xlabel("Time (days)")
    ax_main.set_ylabel("Mean Well-being (0-10)")
    ax_main.set_title("Mean Well-being with SMU-WB Regressions", fontweight="medium")
    ax_main.yaxis.grid(True, linestyle="-", alpha=0.3, color="gray")
    ax_main.set_axisbelow(True)
    ax_main.legend(loc="upper right", framealpha=0.95, edgecolor="none")

    inset_colors = [THESIS_COLORS["muted_red"], THESIS_COLORS["sea_green"],
                    THESIS_COLORS["muted_orange"]]
    inset_position = (0.12, 0.64)
    inset_size = (0.16, 0.18)

    time_min, time_max = float(time_index[0]), float(time_index[-1])
    span = time_max - time_min

    def _expand_limits(values):
        v_min, v_max = float(np.min(values)), float(np.max(values))
        if np.isclose(v_min, v_max):
            padding = max(abs(v_min) * 0.1, 1.0)
            return v_min - padding, v_max + padding
        padding = (v_max - v_min) * 0.05
        return v_min - padding, v_max + padding

    max_time_value = float(np.max(time_index))
    percentile_targets = np.array([0.1, 0.5, 0.9]) * max_time_value

    for i, target in enumerate(percentile_targets):
        t_idx = int(np.abs(time_index - target).argmin())
        x = log.smu[t_idx]
        y = log.wellbeing[t_idx]

        mask = np.isfinite(x) & np.isfinite(y)
        if mask.sum() < 2:
            continue
        x, y = x[mask], y[mask]

        slope, intercept = np.polyfit(x, y, 1)
        residuals = y - (slope * x + intercept)
        ss_res = float(np.dot(residuals, residuals))
        ss_tot = float(np.dot(y - y.mean(), y - y.mean()))
        r_squared = 1.0 - ss_res / ss_tot if ss_tot > 0 else float("nan")

        t_normalised = (float(np.clip(target, time_min, time_max)) - time_min) / span if span > 0 else 0.0
        inset_left = inset_position[0] + t_normalised * (1.0 - inset_position[0] - inset_size[0])
        inset_bottom = inset_position[1]

        ax_inset = fig.add_axes([inset_left, inset_bottom, inset_size[0], inset_size[1]])
        colour = inset_colors[i % len(inset_colors)]
        ax_inset.scatter(x, y, alpha=0.5, s=12, color=colour, edgecolor="white", linewidth=0.3)

        x_range = np.linspace(x.min(), x.max(), 100)
        ax_inset.plot(x_range, slope * x_range + intercept,
                      color=THESIS_COLORS["dark_gray"], linewidth=1.5)

        ax_inset.set_xlim(*_expand_limits(x))
        ax_inset.set_ylim(*_expand_limits(y))
        ax_inset.set_xlabel("SMU", fontsize=9)
        ax_inset.set_ylabel("WB", fontsize=9)
        ax_inset.set_title(f"t={time_index[t_idx]}\nSlope={slope:.2f}, $R^2$={r_squared:.2f}",
                           fontsize=9, fontweight="medium", color=colour)
        ax_inset.tick_params(labelsize=7)
        ax_inset.spines["top"].set_visible(False)
        ax_inset.spines["right"].set_visible(False)

        ax_main.axvline(time_index[t_idx], color=colour, linestyle="--", alpha=0.7, linewidth=1.5)

    fig.tight_layout()

    save_path = os.path.join(output_dir, "abm_simpsons_regressions.pdf")
    fig.savefig(save_path, dpi=300, bbox_inches="tight", facecolor="white")
    print(f"Saved: {save_path}")
    plt.show()
    return fig


# Generate both Simpson's Paradox figures
fig_trends = plot_simpsons_population_means(simpsons_log, OUTPUT_DIR)
fig_regressions = plot_simpsons_regression_overlay(simpsons_log, OUTPUT_DIR)
```
