## Key implementation details

- **Concentrated likelihood.** Both ML estimators concentrate out the linear parameters and $\sigma^2$, reducing the optimisation to a 1D search over $\beta_1$ (SMU) or $\gamma_3$ (WB) via `scipy.optimize.minimize_scalar` on $(-0.99,\,0.99)$. See Section~\ref{sec:est:screen_ml} and Section~\ref{sec:est:wb_ml}.
- **Analytical observed information matrix** for standard errors, built elementwise in the $(\sigma^2)$-parameterisation -- matching the Hessian derivations in Section~\ref{sec:est:screen_hessian} and Section~\ref{sec:est:wb_hessian}. The Hessian-verification cells compare this against a four-point central-difference numerical Hessian.
- **2SLS with HC1-robust standard errors** (sandwich form) for the WB equation. Instruments follow `eq:instrument_set`: $Z = [\mathbf{1}, s, c, Gc, G^2 s, G^2 c]$. First-stage $F$-statistic is the partial $F$ on the three excluded instruments. Sargan-Hansen $J$-test in its Sargan form (normalised by residual variance).
- **Erdos--Renyi networks with no-isolates condition.** `generate_network` redraws until every node has degree $\geq 1$, so $n$ is fixed exactly (matches Assumption~\ref{ass:no_isolates}).
- **A new network is drawn in every replication**, so Monte Carlo variation reflects both network structure and error draws.
- **Parallelisation** via `joblib` with per-replication seeds spawned from a single `SeedSequence` for reproducibility.

## Core Functions

Network generation, data simulation, and estimation routines used throughout the chapter. All estimators return both point estimates and analytical standard errors in the $\sigma^2$ parameterisation, to match the thesis tables.

```python
# ============================================================================
# NETWORK GENERATION AND DATA SIMULATION
# ============================================================================

def generate_network(n, p, rng=None):
    """Generate an Erdos-Renyi G(n, p) network with a row-normalized adjacency matrix.

    Redraws the network until every node has at least one neighbor, so that n
    is fixed exactly (Assumption ``ass:no_isolates``). The returned G is
    row-stochastic: (G @ x)_i is the mean of x over node i's neighbors.
    """
    if rng is None:
        rng = np.random.default_rng()
    while True:
        upper = rng.random((n, n)) < p
        upper = np.triu(upper, k=1)
        adjacency = upper | upper.T
        degrees = adjacency.sum(axis=1)
        if np.all(degrees > 0):
            break

    G = np.zeros((n, n))
    for i in range(n):
        neighbors = np.where(adjacency[i])[0]
        G[i, neighbors] = 1.0 / len(neighbors)
    return G


def simulate_como_data(G, beta0, beta1, sigma_u, gamma0, gamma1, gamma2, gamma3, sigma_eps, rng=None):
    """Simulate one dataset from the recursive COMO model of Chapter 3/5.

    Implements the reduced-form equations from Section ``sec:mc:dgp``:
        s = (I - beta1 G)^{-1} (beta0 1 + u),   u ~ N(0, sigma_u^2 I)
        c = max(0, G s - s)
        y = (I - gamma3 G)^{-1} (gamma0 1 + gamma1 s + gamma2 c + eps),
                                            eps ~ N(0, sigma_eps^2 I)

    Note sigma_u and sigma_eps here are standard deviations; the thesis tables
    report sigma^2, which is handled by the TRUE_SQ dict below.
    """
    if rng is None:
        rng = np.random.default_rng()
    n = G.shape[0]
    I = np.eye(n)
    ones = np.ones(n)

    u = rng.normal(0, sigma_u, n)
    A_s = I - beta1 * G
    s = np.linalg.solve(A_s, beta0 * ones + u)

    s_bar = G @ s
    c = np.maximum(0, s_bar - s)

    eps = rng.normal(0, sigma_eps, n)
    A_y = I - gamma3 * G
    y = np.linalg.solve(A_y, gamma0 * ones + gamma1 * s + gamma2 * c + eps)

    return s, y
```

```python
# ============================================================================
# ML ESTIMATION: SMU EQUATION
# ============================================================================

def estimate_screen_ml(s, G):
    """Concentrated ML for the SMU equation (Algorithm ``alg:screen_ml``).

    Optimisation strategy
    ---------------------
    Concentrate out beta0 (sample mean of A s) and sigma_u^2 (residual variance),
    leaving a 1D search for beta1 in (-0.99, 0.99) via ``minimize_scalar``.
    The concentrated negative log-likelihood is
        -log|det(I - beta1 G)| + (n/2) log(SSE/n).

    Standard errors
    ---------------
    Built from the analytical observed information matrix in the
    (beta0, beta1, sigma_u^2) parameterisation -- see Section
    ``sec:est:screen_hessian``. The information matrix is verified numerically
    in `tab:hessian_check_smu`.
    """
    n = len(s)
    I = np.eye(n)
    ones = np.ones(n)

    def neg_conc_loglik(beta1):
        if abs(beta1) >= 0.99:
            return np.inf
        A = I - beta1 * G
        As = A @ s
        beta0_hat = As.mean()
        resid = As - beta0_hat
        sse = float(resid @ resid)
        sign, log_det = slogdet(A)
        if sign <= 0 or sse <= 0:
            return np.inf
        return -log_det + 0.5 * n * np.log(sse / n)

    res = minimize_scalar(neg_conc_loglik, bounds=(-0.99, 0.99), method='bounded',
                          options={'xatol': 1e-8})

    beta1_hat = res.x
    A = I - beta1_hat * G
    As = A @ s
    beta0_hat = As.mean()
    resid = As - beta0_hat
    sse = float(resid @ resid)
    sigma_u2_hat = sse / n

    # Analytical observed information matrix for (beta0, beta1, sigma_u^2).
    # Elements follow the derivation in Section sec:est:screen_hessian.
    Gs = G @ s
    try:
        A_inv = np.linalg.inv(A)
    except np.linalg.LinAlgError:
        return {'beta0': beta0_hat, 'beta1': beta1_hat,
                'sigma_u_sq': sigma_u2_hat,
                'se_beta0': np.nan, 'se_beta1': np.nan,
                'se_sigma_u_sq': np.nan,
                'success': res.success, 'hessian': None}

    AG = A_inv @ G
    sig2 = sigma_u2_hat
    sig2_sq = sig2**2

    info = np.zeros((3, 3))
    info[0, 0] = n / sig2
    info[0, 1] = info[1, 0] = np.sum(Gs) / sig2
    info[0, 2] = info[2, 0] = np.sum(resid) / sig2_sq
    info[1, 1] = np.trace(AG @ AG) + (Gs @ Gs) / sig2
    info[1, 2] = info[2, 1] = (Gs @ resid) / sig2_sq
    info[2, 2] = -n / (2 * sig2_sq) + sse / (sig2**3)

    try:
        cov_var = np.linalg.inv(info)
        se = np.sqrt(np.maximum(np.diag(cov_var), 0))
    except np.linalg.LinAlgError:
        se = np.array([np.nan, np.nan, np.nan])

    return {'beta0': beta0_hat, 'beta1': beta1_hat,
            'sigma_u_sq': sigma_u2_hat,
            'se_beta0': se[0], 'se_beta1': se[1],
            'se_sigma_u_sq': se[2],
            'success': res.success, 'hessian': info}
```

### Observed information matrix: WB equation

The WB estimator concentrates out the linear block $(\gamma_0, \gamma_1, \gamma_2)$ via OLS on $(I - \gamma_3 G) y$ against $X = [\mathbf{1}, s, c]$, and concentrates out $\sigma_\varepsilon^2$ as the residual variance, leaving a 1D search for $\gamma_3$. Standard errors are built from the analytical $5 \times 5$ observed information matrix in the $(\gamma_0, \gamma_1, \gamma_2, \gamma_3, \sigma_\varepsilon^2)$ parameterisation, filled in element by element below. The derivation is in Section~\ref{sec:est:wb_hessian}; the closed-form matrix is verified numerically against a four-point central difference in `tab:hessian_check_wb`. The two terms that couple $\gamma_3$ to the other parameters come from the Jacobian of $A_y = I - \gamma_3 G$: the $\gamma_3\gamma_3$ entry picks up $\mathrm{tr}(A_y^{-1}G\,A_y^{-1}G)$.

```python
# ============================================================================
# ML ESTIMATION: WELL-BEING EQUATION
# ============================================================================

def estimate_wb_ml(y, s, G):
    """Concentrated ML for the WB equation (Algorithm ``alg:wb_ml``).

    Concentrates out (gamma0, gamma1, gamma2) by OLS of (I - gamma3 G) y on
    X = [1, s, c], and sigma_eps^2 as the residual variance; a 1D search
    in (-0.99, 0.99) recovers gamma3. Returns the analytical observed
    information matrix in the (gamma0, gamma1, gamma2, gamma3, sigma_eps^2)
    parameterisation (Section sec:est:wb_hessian).
    """
    n = len(y)
    I = np.eye(n)
    ones = np.ones(n)
    s_bar = G @ s
    c = np.maximum(0, s_bar - s)
    X = np.column_stack([ones, s, c])

    def neg_conc_loglik(gamma3):
        if abs(gamma3) >= 0.99:
            return np.inf
        A = I - gamma3 * G
        Ay = A @ y
        coef, *_ = lstsq(X, Ay, rcond=None)
        resid = Ay - X @ coef
        sse = float(resid @ resid)
        sign, log_det = slogdet(A)
        if sign <= 0 or sse <= 0:
            return np.inf
        return -log_det + 0.5 * n * np.log(sse / n)

    res = minimize_scalar(neg_conc_loglik, bounds=(-0.99, 0.99), method='bounded',
                          options={'xatol': 1e-8})

    gamma3_hat = res.x
    A = I - gamma3_hat * G
    Ay = A @ y
    coef, *_ = lstsq(X, Ay, rcond=None)
    gamma0_hat, gamma1_hat, gamma2_hat = coef
    resid = Ay - X @ coef
    sse = float(resid @ resid)
    sigma_eps2_hat = sse / n

    # Analytical observed information matrix for
    # (gamma0, gamma1, gamma2, gamma3, sigma_eps^2) -- see sec:est:wb_hessian.
    Gy = G @ y
    try:
        A_inv = np.linalg.inv(A)
    except np.linalg.LinAlgError:
        return {'gamma0': gamma0_hat, 'gamma1': gamma1_hat, 'gamma2': gamma2_hat,
                'gamma3': gamma3_hat, 'sigma_eps_sq': sigma_eps2_hat,
                'se_gamma0': np.nan, 'se_gamma1': np.nan, 'se_gamma2': np.nan,
                'se_gamma3': np.nan, 'se_sigma_eps_sq': np.nan,
                'success': res.success, 'hessian': None}

    AG = A_inv @ G
    sig2 = sigma_eps2_hat
    sig2_sq = sig2**2

    info = np.zeros((5, 5))
    # Linear block: (gamma0, gamma1, gamma2) cross terms (X' X / sig^2)
    info[0, 0] = n / sig2
    info[0, 1] = info[1, 0] = np.sum(s) / sig2
    info[0, 2] = info[2, 0] = np.sum(c) / sig2
    info[1, 1] = np.sum(s**2) / sig2
    info[1, 2] = info[2, 1] = np.sum(s * c) / sig2
    info[2, 2] = np.sum(c**2) / sig2
    # Cross terms with gamma3 (via G y)
    info[0, 3] = info[3, 0] = np.sum(Gy) / sig2
    info[1, 3] = info[3, 1] = (s @ Gy) / sig2
    info[2, 3] = info[3, 2] = (c @ Gy) / sig2
    # gamma3 diagonal: Jacobian trace plus (Gy)'(Gy)/sig^2
    info[3, 3] = np.trace(AG @ AG) + (Gy @ Gy) / sig2
    # Cross terms with sigma_eps^2
    info[0, 4] = info[4, 0] = np.sum(resid) / sig2_sq
    info[1, 4] = info[4, 1] = (s @ resid) / sig2_sq
    info[2, 4] = info[4, 2] = (c @ resid) / sig2_sq
    info[3, 4] = info[4, 3] = (Gy @ resid) / sig2_sq
    info[4, 4] = -n / (2 * sig2_sq) + sse / (sig2**3)

    try:
        cov_var = np.linalg.inv(info)
        se = np.sqrt(np.maximum(np.diag(cov_var), 0))
    except np.linalg.LinAlgError:
        se = np.full(5, np.nan)

    return {'gamma0': gamma0_hat, 'gamma1': gamma1_hat, 'gamma2': gamma2_hat,
            'gamma3': gamma3_hat, 'sigma_eps_sq': sigma_eps2_hat,
            'se_gamma0': se[0], 'se_gamma1': se[1], 'se_gamma2': se[2],
            'se_gamma3': se[3], 'se_sigma_eps_sq': se[4],
            'success': res.success, 'hessian': info}
```

### 2SLS estimator for the WB equation

Implements the 2SLS procedure from Section~\ref{sec:est:wb_2sls} with the instrument set $Z = [\mathbf{1}, s, c, Gc, G^2 s, G^2 c]$ from `eq:instrument_set`. The endogenous regressor is $Gy$. Standard errors use the HC1 heteroskedasticity-robust sandwich with the standard $n/(n-k)$ degrees-of-freedom correction.

The function also returns:
- The **first-stage partial $F$-statistic** on the three excluded instruments ($Gc$, $G^2 s$, $G^2 c$), compared in the thesis to the Stock-Yogo/Staiger-Stock threshold of 10.
- The **Sargan-Hansen $J$-test** in Sargan form (normalised by residual variance), distributed as $\chi^2_{l-k} = \chi^2_2$ under the null of valid instruments.
- The **Hausman test** against OLS on $\gamma_3$. The variance difference $\widehat{\mathrm{Var}}(\hat\gamma_3^{\text{2SLS}}) - \widehat{\mathrm{Var}}(\hat\gamma_3^{\text{OLS}})$ can go negative in finite samples; when that happens the test statistic is reported as NaN (see the Remark in Section~\ref{sec:mc:overid}).
- The full OLS estimates, reused for `tab:mc_ols_bias`.

```python
# ============================================================================
# 2SLS ESTIMATION: WELL-BEING EQUATION
# ============================================================================

def estimate_wb_2sls(y, s, G):
    """2SLS for the WB equation with network instruments (Section sec:est:wb_2sls).

    Regressors W = [1, s, c, Gy], endogenous = Gy.
    Instruments  Z = [1, s, c, Gc, G^2 s, G^2 c]  (eq:instrument_set).
    Standard errors: HC1 robust sandwich.

    Also returns the first-stage partial F on the excluded instruments,
    the Sargan-Hansen J-statistic (Sargan form) and its p-value,
    the Hausman statistic/p-value on gamma3 (NaN if the variance difference
    is non-positive, as discussed in the Remark in sec:mc:overid), and the
    OLS coefficients (reused for the OLS bias table).
    """
    n = len(y)
    ones = np.ones(n)
    s_bar = G @ s
    c = np.maximum(0, s_bar - s)
    y_bar = G @ y  # endogenous regressor

    G2 = G @ G

    # Regressors: [1, s, c, Gy]
    W = np.column_stack([ones, s, c, y_bar])
    k = W.shape[1]

    # Full instrument set Z from eq:instrument_set
    Z = np.column_stack([ones, s, c, G @ c, G2 @ s, G2 @ c])
    l = Z.shape[1]

    # Included (exogenous) instruments for the partial F-test
    Z_included = np.column_stack([ones, s, c])
    q_excluded = l - Z_included.shape[1]  # = 3 excluded instruments

    # --- First stage ---
    pi_hat = lstsq(Z, y_bar, rcond=None)[0]
    y_bar_fitted = Z @ pi_hat
    first_stage_resid = y_bar - y_bar_fitted

    # First-stage partial F-statistic on the excluded instruments
    pi_r = lstsq(Z_included, y_bar, rcond=None)[0]
    ssr_restricted = np.sum((y_bar - Z_included @ pi_r)**2)
    ssr_unrestricted = np.sum(first_stage_resid**2)
    f_stat = ((ssr_restricted - ssr_unrestricted) / q_excluded) / (ssr_unrestricted / max(n - l, 1))

    # --- Second stage ---
    W_hat = W.copy()
    W_hat[:, 3] = y_bar_fitted  # replace endogenous with fitted
    beta_2sls = lstsq(W_hat, y, rcond=None)[0]

    # Residuals evaluated at the original (not fitted) W
    resid = y - W @ beta_2sls
    sigma2_hat = np.sum(resid**2) / (n - k)

    # --- HC1 sandwich standard errors ---
    # (W' P_Z W)^{-1} * [n/(n-k) * sum_i e_i^2 (P_Z W)_i (P_Z W)_i'] * (W' P_Z W)^{-1}
    try:
        ZtZ_inv = inv(Z.T @ Z)
        P_Z = Z @ ZtZ_inv @ Z.T
        WtPzW = W.T @ P_Z @ W
        WtPzW_inv = inv(WtPzW)
        PzW = P_Z @ W
        meat = np.zeros((k, k))
        for i in range(n):
            pzw_i = PzW[i, :]
            meat += resid[i]**2 * np.outer(pzw_i, pzw_i)
        meat *= n / (n - k)  # HC1 degrees-of-freedom correction
        cov_robust = WtPzW_inv @ meat @ WtPzW_inv
        se = np.sqrt(np.maximum(np.diag(cov_robust), 0))
    except np.linalg.LinAlgError:
        se = np.full(k, np.nan)
        P_Z = None
        cov_robust = None

    # --- Sargan-Hansen J-test (Sargan form, normalised by residual variance) ---
    if l > k:
        try:
            ete = resid @ resid
            j_stat = float(n * (resid @ Z @ ZtZ_inv @ Z.T @ resid) / ete)
            j_pval = 1 - stats.chi2.cdf(j_stat, l - k)
        except Exception:
            j_stat, j_pval = np.nan, np.nan
    else:
        j_stat, j_pval = np.nan, np.nan

    # --- OLS (used for the Hausman test and tab:mc_ols_bias) ---
    beta_ols = lstsq(W, y, rcond=None)[0]
    resid_ols = y - W @ beta_ols
    sigma2_ols = np.sum(resid_ols**2) / (n - k)
    try:
        cov_ols = sigma2_ols * inv(W.T @ W)
        # Hausman test on gamma3 (index 3). The variance difference can be
        # negative in finite samples; in that case the statistic is NaN.
        var_diff = cov_robust[3, 3] - cov_ols[3, 3]
        if var_diff > 0:
            h_stat = (beta_2sls[3] - beta_ols[3])**2 / var_diff
            h_pval = 1 - stats.chi2.cdf(h_stat, 1)
        else:
            h_stat, h_pval = np.nan, np.nan
    except Exception:
        h_stat, h_pval = np.nan, np.nan
        beta_ols = np.full(k, np.nan)

    return {
        'gamma0': beta_2sls[0], 'gamma1': beta_2sls[1],
        'gamma2': beta_2sls[2], 'gamma3': beta_2sls[3],
        'se_gamma0': se[0], 'se_gamma1': se[1],
        'se_gamma2': se[2], 'se_gamma3': se[3],
        'f_stat': f_stat, 'j_stat': j_stat, 'j_pval': j_pval,
        'h_stat': h_stat, 'h_pval': h_pval,
        'cov_robust': cov_robust,
        'ols_gamma0': beta_ols[0], 'ols_gamma1': beta_ols[1],
        'ols_gamma2': beta_ols[2], 'ols_gamma3': beta_ols[3],
        'success': True,
    }
```

```python
# ============================================================================
# NUMERICAL HESSIAN (verification against analytical information matrix)
# ============================================================================

def full_loglik_wb(params, y, s, G):
    """Full (non-concentrated) negative log-likelihood for the WB equation
    in the (gamma0, gamma1, gamma2, gamma3, sigma_eps^2) parameterisation.
    Used only as input to `numerical_hessian` for tab:hessian_check_wb."""
    gamma0, gamma1, gamma2, gamma3, sigma_eps2 = params
    if abs(gamma3) >= 0.99 or sigma_eps2 <= 0:
        return np.inf
    n = len(y)
    I = np.eye(n)
    A = I - gamma3 * G
    s_bar = G @ s
    c = np.maximum(0, s_bar - s)
    X = np.column_stack([np.ones(n), s, c])
    Ay = A @ y
    r = Ay - X @ np.array([gamma0, gamma1, gamma2])
    sse = float(r @ r)
    sign, log_det = slogdet(A)
    if sign <= 0:
        return np.inf
    return -log_det + 0.5 * n * np.log(sigma_eps2) + sse / (2 * sigma_eps2)


def full_loglik_screen(params, s, G):
    """Full negative log-likelihood for the SMU equation in the
    (beta0, beta1, sigma_u^2) parameterisation. Used for tab:hessian_check_smu."""
    beta0, beta1, sigma_u2 = params
    if abs(beta1) >= 0.99 or sigma_u2 <= 0:
        return np.inf
    n = len(s)
    I = np.eye(n)
    A = I - beta1 * G
    As = A @ s
    r = As - beta0 * np.ones(n)
    sse = float(r @ r)
    sign, log_det = slogdet(A)
    if sign <= 0:
        return np.inf
    return -log_det + 0.5 * n * np.log(sigma_u2) + sse / (2 * sigma_u2)


def numerical_hessian(func, params, eps=1e-5):
    """Four-point central-difference approximation to the Hessian of `func`.

    Matches the formula in Section ``sec:mc:hessian`` with step h = 1e-5.
    """
    p = len(params)
    H = np.zeros((p, p))
    for i in range(p):
        for j in range(i, p):
            pp = params.copy(); pp[i] += eps; pp[j] += eps
            pm = params.copy(); pm[i] += eps; pm[j] -= eps
            mp = params.copy(); mp[i] -= eps; mp[j] += eps
            mm = params.copy(); mm[i] -= eps; mm[j] -= eps
            H[i, j] = (func(pp) - func(pm) - func(mp) + func(mm)) / (4 * eps**2)
            H[j, i] = H[i, j]
    return H
```

## Section 6.3--6.5: Main Monte Carlo Study (Tables 6.4--6.11, Figures)

```python
def run_main_replication(n, p, true, seed):
    """Run one replication of the main Monte Carlo study.

    Draws a fresh network and dataset, then applies all three estimators
    (SMU ML, WB ML, WB 2SLS). Returns a dict of point estimates, standard
    errors, and 2SLS diagnostics. Failures silently set fields to NaN.
    """
    smu_params = ['beta0', 'beta1', 'sigma_u_sq']
    wb_params = ['gamma0', 'gamma1', 'gamma2', 'gamma3', 'sigma_eps_sq']
    wb_params_iv = ['gamma0', 'gamma1', 'gamma2', 'gamma3']

    out = {
        'ml_screen': {k: np.nan for k in smu_params},
        'ml_screen_se': {k: np.nan for k in smu_params},
        'ml_wb': {k: np.nan for k in wb_params},
        'ml_wb_se': {k: np.nan for k in wb_params},
        'iv_wb': {k: np.nan for k in wb_params_iv},
        'iv_wb_se': {k: np.nan for k in wb_params_iv},
        'ols_wb': {k: np.nan for k in wb_params_iv},
        'f_stat': np.nan,
        'j_stat': np.nan,
        'j_pval': np.nan,
        'h_stat': np.nan,
        'h_pval': np.nan,
    }

    try:
        rng = np.random.default_rng(seed)

        # A fresh network and dataset for every replication (Section sec:mc:procedure).
        G = generate_network(n, p, rng)
        s, y = simulate_como_data(G, **true, rng=rng)

        # SMU ML
        res_s = estimate_screen_ml(s, G)
        for k in smu_params:
            out['ml_screen'][k] = res_s[k]
            out['ml_screen_se'][k] = res_s[f'se_{k}']

        # WB ML
        res_w = estimate_wb_ml(y, s, G)
        for k in wb_params:
            out['ml_wb'][k] = res_w[k]
            out['ml_wb_se'][k] = res_w[f'se_{k}']

        # WB 2SLS (and OLS from inside the same call)
        res_iv = estimate_wb_2sls(y, s, G)
        for k in wb_params_iv:
            out['iv_wb'][k] = res_iv[k]
            out['iv_wb_se'][k] = res_iv[f'se_{k}']
        for k in wb_params_iv:
            out['ols_wb'][k] = res_iv[f'ols_{k}']

        out['f_stat'] = res_iv['f_stat']
        out['j_stat'] = res_iv['j_stat']
        out['j_pval'] = res_iv['j_pval']
        out['h_stat'] = res_iv['h_stat']
        out['h_pval'] = res_iv['h_pval']

    except Exception:
        pass

    return out


def _aggregate_main_replications(rep_results):
    """Stack per-replication dicts into arrays keyed by parameter."""
    smu_params = ['beta0', 'beta1', 'sigma_u_sq']
    wb_params = ['gamma0', 'gamma1', 'gamma2', 'gamma3', 'sigma_eps_sq']
    wb_params_iv = ['gamma0', 'gamma1', 'gamma2', 'gamma3']

    ml_screen = {k: np.array([rr['ml_screen'][k] for rr in rep_results], dtype=float) for k in smu_params}
    ml_screen_se = {k: np.array([rr['ml_screen_se'][k] for rr in rep_results], dtype=float) for k in smu_params}

    ml_wb = {k: np.array([rr['ml_wb'][k] for rr in rep_results], dtype=float) for k in wb_params}
    ml_wb_se = {k: np.array([rr['ml_wb_se'][k] for rr in rep_results], dtype=float) for k in wb_params}

    iv_wb = {k: np.array([rr['iv_wb'][k] for rr in rep_results], dtype=float) for k in wb_params_iv}
    iv_wb_se = {k: np.array([rr['iv_wb_se'][k] for rr in rep_results], dtype=float) for k in wb_params_iv}

    ols_wb = {k: np.array([rr['ols_wb'][k] for rr in rep_results], dtype=float) for k in wb_params_iv}

    return {
        'ml_screen': ml_screen,
        'ml_screen_se': ml_screen_se,
        'ml_wb': ml_wb,
        'ml_wb_se': ml_wb_se,
        'iv_wb': iv_wb,
        'iv_wb_se': iv_wb_se,
        'ols_wb': ols_wb,
        'f_stats': np.array([rr['f_stat'] for rr in rep_results], dtype=float),
        'j_stats': np.array([rr['j_stat'] for rr in rep_results], dtype=float),
        'j_pvals': np.array([rr['j_pval'] for rr in rep_results], dtype=float),
        'h_stats': np.array([rr['h_stat'] for rr in rep_results], dtype=float),
        'h_pvals': np.array([rr['h_pval'] for rr in rep_results], dtype=float),
    }


def run_main_mc_parallel(n, p, R, true, seed=42, n_jobs=None):
    """Run the main Monte Carlo study in parallel using joblib.

    Seeds are spawned from a single SeedSequence so the run is reproducible
    and each replication is independent.
    """
    seed_seq = np.random.SeedSequence(seed)
    child_seqs = seed_seq.spawn(R)
    rep_seeds = [int(cs.generate_state(1, dtype=np.uint32)[0]) for cs in child_seqs]

    if n_jobs is None:
        n_jobs = -1

    with parallel_config(backend='loky', inner_max_num_threads=1):
        rep_results = Parallel(n_jobs=n_jobs)(
            delayed(run_main_replication)(n, p, true, rep_seed)
            for rep_seed in rep_seeds
        )

    return _aggregate_main_replications(rep_results)
```

```python
print(f"Running main Monte Carlo study: n={N_BASE}, p={P_BASE}, R={R_MAIN}")
print("This will take several minutes...")
mc = run_main_mc_parallel(N_BASE, P_BASE, R_MAIN, TRUE, seed=RNG_SEED, n_jobs=-1)
print("Done.")
```

```python
def mc_summary(est, se, true_val, name):
    """Compute MC summary statistics for one parameter."""
    mask = np.isfinite(est) & np.isfinite(se)
    e, s = est[mask], se[mask]
    mean_est = np.mean(e)
    bias = mean_est - true_val
    std_est = np.std(e, ddof=1)
    rmse = np.sqrt(bias**2 + std_est**2)
    mean_se = np.mean(s)
    # Coverage
    lower = e - 1.96 * s
    upper = e + 1.96 * s
    coverage = np.mean((lower <= true_val) & (true_val <= upper))
    return {'Parameter': name, 'True': true_val, 'Mean Est.': mean_est,
            'Bias': bias, 'Std(Est.)': std_est, 'Mean SE': mean_se,
            'RMSE': rmse, 'Coverage': coverage, 'N_valid': mask.sum()}
```

## Section 6.5: Wald Test Size and Power (Table 6.12)

We now validate the Wald-type tests described in Section~\ref{sec:est:testing}
by measuring their empirical size and power under two hypotheses:

- $H_0^{(a)}$: $\gamma_2 = 0$ (no CoMO effect; single Wald $t$-test)
- $H_0^{(b)}$: $\gamma_2 = \gamma_3 = 0$ (no CoMO effect and no network peer effect; joint Wald $\chi^2_2$ test)

Size is the rejection rate at $\alpha = 0.05$ under a joint null DGP with both
$\gamma_2$ and $\gamma_3$ set to zero. The same DGP is used to size both tests:
for $H_0^{(b)}$ this is the full null, while for $H_0^{(a)}$ it fixes the nuisance
parameter $\gamma_3$ at zero rather than scanning across the stability region.
Power is the rejection rate under the main DGP (`TRUE` with $\gamma_2 = -0.8$,
$\gamma_3 = 0.25$) and saturates at $1.000$, reflecting a deliberately
well-separated alternative rather than a stringent test of the power function.
Both ML (observed information matrix) and 2SLS (HC1-robust covariance) variance
estimates are evaluated. We use $R = 1{,}000$ replications at $n = 500$ and $p = 0.01$.


```python
# --- Wald test size and power: replication and aggregation ---

def run_wald_replication(n, p, true, seed):
    """One replication: returns p-values of Wald tests for gamma_2 = 0 and
    the joint hypothesis gamma_2 = gamma_3 = 0 under both ML and 2SLS.

    The ML covariance is obtained by inverting the analytical observed
    information matrix (indices 2, 3 in the (gamma0, gamma1, gamma2, gamma3,
    sigma_eps^2) parameterisation). The 2SLS covariance is the HC1-robust
    sandwich matrix returned by estimate_wb_2sls (indices 2, 3 in the
    (gamma0, gamma1, gamma2, gamma3) parameterisation).

    Single-parameter test: z = gamma2_hat / sqrt(Var(gamma2_hat)) -> N(0, 1).
    Joint test: W = theta' Sigma^{-1} theta -> chi^2_2.
    """
    out = {'ml_p_single': np.nan, 'ml_p_joint': np.nan,
           'iv_p_single': np.nan, 'iv_p_joint': np.nan}
    try:
        rng = np.random.default_rng(seed)
        G = generate_network(n, p, rng)
        s, y = simulate_como_data(G, rng=rng, **true)

        # ML
        rw = estimate_wb_ml(y, s, G)
        if rw.get('hessian') is not None:
            try:
                cov = np.linalg.inv(rw['hessian'])
                g2, g3 = rw['gamma2'], rw['gamma3']
                var2 = max(cov[2, 2], 0.0)
                if var2 > 0:
                    z = g2 / np.sqrt(var2)
                    out['ml_p_single'] = 2 * (1 - stats.norm.cdf(abs(z)))
                sub = cov[np.ix_([2, 3], [2, 3])]
                vec = np.array([g2, g3])
                W = float(vec @ np.linalg.solve(sub, vec))
                out['ml_p_joint'] = float(1 - stats.chi2.cdf(W, 2))
            except np.linalg.LinAlgError:
                pass

        # 2SLS
        ri = estimate_wb_2sls(y, s, G)
        if ri.get('cov_robust') is not None:
            try:
                cr = ri['cov_robust']
                g2i, g3i = ri['gamma2'], ri['gamma3']
                var2i = max(cr[2, 2], 0.0)
                if var2i > 0:
                    z = g2i / np.sqrt(var2i)
                    out['iv_p_single'] = 2 * (1 - stats.norm.cdf(abs(z)))
                sub = cr[np.ix_([2, 3], [2, 3])]
                vec = np.array([g2i, g3i])
                W = float(vec @ np.linalg.solve(sub, vec))
                out['iv_p_joint'] = float(1 - stats.chi2.cdf(W, 2))
            except np.linalg.LinAlgError:
                pass
    except Exception:
        pass
    return out


def run_wald_mc_parallel(n, p, R, true, seed, n_jobs=None):
    """Run the Wald-test Monte Carlo in parallel using joblib.

    Mirrors run_main_mc_parallel: spawns R independent seeds from a single
    SeedSequence so the run is reproducible.
    """
    if n_jobs is None:
        n_jobs = -1
    seed_seq = np.random.SeedSequence(seed)
    child_seqs = seed_seq.spawn(R)
    rep_seeds = [int(cs.generate_state(1, dtype=np.uint32)[0]) for cs in child_seqs]

    with parallel_config(backend='loky', inner_max_num_threads=1):
        rep_results = Parallel(n_jobs=n_jobs)(
            delayed(run_wald_replication)(n, p, true, rs) for rs in rep_seeds
        )
    keys = ['ml_p_single', 'ml_p_joint', 'iv_p_single', 'iv_p_joint']
    return {k: np.array([rr[k] for rr in rep_results], dtype=float) for k in keys}


# Null DGP: identical to TRUE except gamma2 = gamma3 = 0 (size study)
TRUE_NULL = dict(TRUE)
TRUE_NULL['gamma2'] = 0.0
TRUE_NULL['gamma3'] = 0.0

print(f"Running Wald size MC (null DGP, gamma2=gamma3=0): n={N_BASE}, p={P_BASE}, R={R_MAIN}")
wald_size = run_wald_mc_parallel(N_BASE, P_BASE, R_MAIN, TRUE_NULL, seed=RNG_SEED, n_jobs=-1)

print(f"Running Wald power MC (main DGP): n={N_BASE}, p={P_BASE}, R={R_MAIN}")
wald_power = run_wald_mc_parallel(N_BASE, P_BASE, R_MAIN, TRUE, seed=RNG_SEED + 1, n_jobs=-1)
```

```python
# --- Table 6.12: Wald-test size and power ---
ALPHA = 0.05

def _rej(pvals, alpha=ALPHA):
    m = np.isfinite(pvals)
    if not m.any():
        return np.nan
    return float(np.mean(pvals[m] < alpha))

rows = []
for label, key_single, key_joint in [
    ('ML',   'ml_p_single', 'ml_p_joint'),
    ('2SLS', 'iv_p_single', 'iv_p_joint'),
]:
    for test_name, key in [
        ('Single  (H0: γ₂ = 0)',             key_single),
        ('Joint   (H0: γ₂ = γ₃ = 0)', key_joint),
    ]:
        size_rate = _rej(wald_size[key])
        power_rate = _rej(wald_power[key])
        rows.append((label, test_name, size_rate, power_rate))

print(f"Table 6.12: Wald-test size and power (n={N_BASE}, R={R_MAIN}, α={ALPHA})")
print("=" * 70)
print(f"{'Estimator':<8} {'Test':<34} {'Size':>8} {'Power':>8}")
print("-" * 70)
for r in rows:
    print(f"{r[0]:<8} {r[1]:<34} {r[2]:>8.3f} {r[3]:>8.3f}")
print("=" * 70)
print(f"Note: size is the rejection rate under the null DGP (γ₂ = γ₃ = 0);")
print(f"power is the rejection rate under the main DGP (γ₂ = -0.8, γ₃ = 0.25).")
```

```python
# Figure: empirical CDFs of joint Wald p-values under the null DGP.
# Under a correctly sized test these should lie on the 45-degree line.
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

_panels = [
    ('ml_p_joint',  'ML joint Wald $p$-values under $H_0: \\gamma_2 = \\gamma_3 = 0$'),
    ('iv_p_joint',  '2SLS joint Wald $p$-values under $H_0: \\gamma_2 = \\gamma_3 = 0$'),
]
for ax, (key, title) in zip(axes, _panels):
    pv = wald_size[key]
    pv = pv[np.isfinite(pv)]
    pv_sorted = np.sort(pv)
    grid = np.arange(1, len(pv_sorted) + 1) / len(pv_sorted)
    ax.plot(pv_sorted, grid, lw=1.5, label='Empirical CDF')
    ax.plot([0, 1], [0, 1], 'r--', lw=1, label='Uniform[0,1]')
    ax.axvline(ALPHA, color='grey', linestyle=':', lw=1, alpha=0.7,
               label=f'$\\alpha = {ALPHA}$')
    ax.set_xlabel('$p$-value')
    ax.set_ylabel('Empirical CDF')
    ax.set_title(title, fontsize=10)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.legend(loc='lower right', fontsize=9)

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/mc_wald_pvalues.pdf', bbox_inches='tight')
plt.show()
```

## Section 6.6: Effect of Network Density (Tables 6.12, 6.13)

```python
def run_density_study(n, densities, R, true, seed=100, n_jobs=-1):
    """Run the main MC across a grid of network densities.

    For each connection probability p, returns three DataFrames with the
    performance metrics for gamma2 and gamma3 (ML and 2SLS side-by-side),
    plus an identification check that verifies the Bramoulle condition
    rank[I, G, G^2] = 3 in every replication and reports the mean
    first-stage F statistic (tab:mc_identification).
    """
    results_g2 = []
    results_g3 = []
    results_id = []
    for p in densities:
        exp_deg = (n - 1) * p
        print(f"  p={p:.3f} (expected degree ~ {exp_deg:.1f})...")
        mc = run_main_mc_parallel(n, p, R, true, seed=seed, n_jobs=n_jobs)

        # --- gamma2 ---
        ml_e = mc['ml_wb']['gamma2']
        ml_se = mc['ml_wb_se']['gamma2']
        mask = np.isfinite(ml_e) & np.isfinite(ml_se)
        ml_bias = np.mean(ml_e[mask]) - true['gamma2']
        ml_rmse = np.sqrt(ml_bias**2 + np.var(ml_e[mask], ddof=1))
        ml_cov = np.mean((ml_e[mask] - 1.96*ml_se[mask] <= true['gamma2']) &
                         (true['gamma2'] <= ml_e[mask] + 1.96*ml_se[mask]))

        iv_e = mc['iv_wb']['gamma2']
        iv_se = mc['iv_wb_se']['gamma2']
        mask_iv = np.isfinite(iv_e) & np.isfinite(iv_se)
        iv_bias = np.mean(iv_e[mask_iv]) - true['gamma2']
        iv_rmse = np.sqrt(iv_bias**2 + np.var(iv_e[mask_iv], ddof=1))
        iv_cov = np.mean((iv_e[mask_iv] - 1.96*iv_se[mask_iv] <= true['gamma2']) &
                         (true['gamma2'] <= iv_e[mask_iv] + 1.96*iv_se[mask_iv]))

        results_g2.append({
            'p': p, 'Exp. Degree': exp_deg,
            'ML Bias': ml_bias, 'ML RMSE': ml_rmse, 'ML Coverage': ml_cov,
            '2SLS Bias': iv_bias, '2SLS RMSE': iv_rmse, '2SLS Coverage': iv_cov,
        })

        # --- gamma3 ---
        ml_e3 = mc['ml_wb']['gamma3']
        ml_se3 = mc['ml_wb_se']['gamma3']
        mask3 = np.isfinite(ml_e3) & np.isfinite(ml_se3)
        ml_bias3 = np.mean(ml_e3[mask3]) - true['gamma3']
        ml_rmse3 = np.sqrt(ml_bias3**2 + np.var(ml_e3[mask3], ddof=1))
        ml_cov3 = np.mean((ml_e3[mask3] - 1.96*ml_se3[mask3] <= true['gamma3']) &
                          (true['gamma3'] <= ml_e3[mask3] + 1.96*ml_se3[mask3]))

        iv_e3 = mc['iv_wb']['gamma3']
        iv_se3 = mc['iv_wb_se']['gamma3']
        mask_iv3 = np.isfinite(iv_e3) & np.isfinite(iv_se3)
        iv_bias3 = np.mean(iv_e3[mask_iv3]) - true['gamma3']
        iv_rmse3 = np.sqrt(iv_bias3**2 + np.var(iv_e3[mask_iv3], ddof=1))
        iv_cov3 = np.mean((iv_e3[mask_iv3] - 1.96*iv_se3[mask_iv3] <= true['gamma3']) &
                          (true['gamma3'] <= iv_e3[mask_iv3] + 1.96*iv_se3[mask_iv3]))

        results_g3.append({
            'p': p, 'Exp. Degree': exp_deg,
            'ML Bias': ml_bias3, 'ML RMSE': ml_rmse3, 'ML Coverage': ml_cov3,
            '2SLS Bias': iv_bias3, '2SLS RMSE': iv_rmse3, '2SLS Coverage': iv_cov3,
        })

        # --- Bramoulle identification check + mean first-stage F ---
        # Re-draw networks with the same seed used inside run_main_mc_parallel
        # so that the rank check is performed on exactly the same networks
        # the estimators saw. rank[I, G, G^2] = 3 is the condition of Theorem
        # thm:bramoulle.
        f = mc['f_stats'][np.isfinite(mc['f_stats'])]
        mean_f = np.mean(f) if len(f) > 0 else np.nan

        seed_seq = np.random.SeedSequence(seed)
        child_seqs = seed_seq.spawn(R)
        rep_seeds = [int(cs.generate_state(1, dtype=np.uint32)[0]) for cs in child_seqs]
        n_identified = 0
        for rep_seed in rep_seeds:
            rng_r = np.random.default_rng(rep_seed)
            G_r = generate_network(n, p, rng_r)
            G2_r = G_r @ G_r
            n_eff = G_r.shape[0]
            M = np.column_stack([np.eye(n_eff).ravel(), G_r.ravel(), G2_r.ravel()])
            if matrix_rank(M) == 3:
                n_identified += 1
        frac_identified = n_identified / R

        results_id.append({
            'p': p, 'Exp. Degree': exp_deg,
            'Mean F': mean_f, 'Frac. Identified': frac_identified,
        })

    return pd.DataFrame(results_g2), pd.DataFrame(results_g3), pd.DataFrame(results_id)
```

```python
densities = [0.01, 0.02, 0.03, 0.05, 0.10]
print(f"Running density study: n={N_BASE}, R={R_SENS}")
df_density_g2, df_density_g3, df_density_id = run_density_study(N_BASE, densities, R_SENS, TRUE, seed=100, n_jobs=-1)

print("\nTable: Effect of network density on gamma2 estimation (n=500, R=500)")
print("=" * 90)
print(df_density_g2[['p', 'Exp. Degree', 'ML Bias', 'ML RMSE', 'ML Coverage',
                   '2SLS Bias', '2SLS RMSE', '2SLS Coverage']].to_string(index=False, float_format='{:.4f}'.format))

print("\nTable: Effect of network density on gamma3 estimation (n=500, R=500)")
print("=" * 90)
print(df_density_g3[['p', 'Exp. Degree', 'ML Bias', 'ML RMSE', 'ML Coverage',
                   '2SLS Bias', '2SLS RMSE', '2SLS Coverage']].to_string(index=False, float_format='{:.4f}'.format))

print("\nTable: Identification conditions across network densities")
print("=" * 50)
print(df_density_id[['p', 'Frac. Identified', 'Mean F']].to_string(
    index=False,
    formatters={'p': '{:.3f}'.format, 'Frac. Identified': '{:.2f}'.format, 'Mean F': '{:.2f}'.format}
))
```

## Section 6.7: Convergence as Sample Size Increases (Tables 6.14--6.16)

```python
def run_convergence_study(sample_sizes, target_degree, R, true, seed=300, n_jobs=-1):
    """Run the main MC across a grid of sample sizes with fixed expected degree.

    For each sample size n, the connection probability is set to
    p = target_degree / (n - 1), so the expected degree stays constant.
    This isolates the effect of increasing n from the effect of changing
    network density (cf. Section sec:mc:density).

    Returns four DataFrames -- ML and 2SLS results for gamma2 and gamma3.
    """
    results_ml_g2 = []
    results_iv_g2 = []
    results_ml_g3 = []
    results_iv_g3 = []

    for n in sample_sizes:
        p = target_degree / (n - 1)
        exp_deg = (n - 1) * p  # == target_degree
        print(f"  n={n}, p={p:.4f} (expected degree = {exp_deg:.1f})...")
        mc = run_main_mc_parallel(n, p, R, true, seed=seed, n_jobs=n_jobs)

        # --- gamma2 (ML) ---
        e = mc['ml_wb']['gamma2']
        se = mc['ml_wb_se']['gamma2']
        mask = np.isfinite(e) & np.isfinite(se)
        bias = np.mean(e[mask]) - true['gamma2']
        std = np.std(e[mask], ddof=1)
        rmse = np.sqrt(bias**2 + std**2)
        cov = np.mean((e[mask] - 1.96*se[mask] <= true['gamma2']) &
                      (true['gamma2'] <= e[mask] + 1.96*se[mask]))
        results_ml_g2.append({'n': n, 'p': round(p, 4), 'Bias': bias,
                              'Std(Est.)': std, 'RMSE': rmse, 'Coverage': cov})

        # --- gamma2 (2SLS) ---
        e_iv = mc['iv_wb']['gamma2']
        se_iv = mc['iv_wb_se']['gamma2']
        mask_iv = np.isfinite(e_iv) & np.isfinite(se_iv)
        bias_iv = np.mean(e_iv[mask_iv]) - true['gamma2']
        std_iv = np.std(e_iv[mask_iv], ddof=1)
        rmse_iv = np.sqrt(bias_iv**2 + std_iv**2)
        cov_iv = np.mean((e_iv[mask_iv] - 1.96*se_iv[mask_iv] <= true['gamma2']) &
                         (true['gamma2'] <= e_iv[mask_iv] + 1.96*se_iv[mask_iv]))
        results_iv_g2.append({'n': n, 'p': round(p, 4), 'Bias': bias_iv,
                              'Std(Est.)': std_iv, 'RMSE': rmse_iv, 'Coverage': cov_iv})

        # --- gamma3 (ML) ---
        e_ml3 = mc['ml_wb']['gamma3']
        se_ml3 = mc['ml_wb_se']['gamma3']
        mask_ml3 = np.isfinite(e_ml3) & np.isfinite(se_ml3)
        bias_ml3 = np.mean(e_ml3[mask_ml3]) - true['gamma3']
        std_ml3 = np.std(e_ml3[mask_ml3], ddof=1)
        rmse_ml3 = np.sqrt(bias_ml3**2 + std_ml3**2)
        cov_ml3 = np.mean((e_ml3[mask_ml3] - 1.96*se_ml3[mask_ml3] <= true['gamma3']) &
                          (true['gamma3'] <= e_ml3[mask_ml3] + 1.96*se_ml3[mask_ml3]))
        results_ml_g3.append({'n': n, 'p': round(p, 4), 'Bias': bias_ml3,
                              'Std(Est.)': std_ml3, 'RMSE': rmse_ml3, 'Coverage': cov_ml3})

        # --- gamma3 (2SLS) ---
        e_iv3 = mc['iv_wb']['gamma3']
        se_iv3 = mc['iv_wb_se']['gamma3']
        mask_iv3 = np.isfinite(e_iv3) & np.isfinite(se_iv3)
        bias_iv3 = np.mean(e_iv3[mask_iv3]) - true['gamma3']
        std_iv3 = np.std(e_iv3[mask_iv3], ddof=1)
        rmse_iv3 = np.sqrt(bias_iv3**2 + std_iv3**2)
        cov_iv3 = np.mean((e_iv3[mask_iv3] - 1.96*se_iv3[mask_iv3] <= true['gamma3']) &
                          (true['gamma3'] <= e_iv3[mask_iv3] + 1.96*se_iv3[mask_iv3]))
        results_iv_g3.append({'n': n, 'p': round(p, 4), 'Bias': bias_iv3,
                              'Std(Est.)': std_iv3, 'RMSE': rmse_iv3, 'Coverage': cov_iv3})

    return (pd.DataFrame(results_ml_g2), pd.DataFrame(results_iv_g2),
            pd.DataFrame(results_ml_g3), pd.DataFrame(results_iv_g3))
```

```python
CONV_DEGREE = 10  # fixed expected degree for the convergence study
sample_sizes = [500, 750, 1000, 1500, 2000]
print(f"Running convergence study: fixed expected degree={CONV_DEGREE}, R={R_SENS}")
df_conv_ml_g2, df_conv_iv_g2, df_conv_ml_g3, df_conv_iv_g3 = run_convergence_study(
    sample_sizes, CONV_DEGREE, R_SENS, TRUE, seed=300, n_jobs=-1)

print(f"\nTable: Convergence of ML estimator for gamma2 (True = -0.8, exp. degree={CONV_DEGREE}, R=500)")
print("=" * 70)
print(df_conv_ml_g2.to_string(index=False, float_format='{:.4f}'.format))

print(f"\nTable: Convergence of 2SLS estimator for gamma2 (True = -0.8, exp. degree={CONV_DEGREE}, R=500)")
print("=" * 70)
print(df_conv_iv_g2.to_string(index=False, float_format='{:.4f}'.format))

print(f"\nTable: Convergence of ML estimator for gamma3 (True = 0.25, exp. degree={CONV_DEGREE}, R=500)")
print("=" * 70)
print(df_conv_ml_g3.to_string(index=False, float_format='{:.4f}'.format))

print(f"\nTable: Convergence of 2SLS estimator for gamma3 (True = 0.25, exp. degree={CONV_DEGREE}, R=500)")
print("=" * 70)
print(df_conv_iv_g3.to_string(index=False, float_format='{:.4f}'.format))
```
